```python
"""
Advanced Hybrid RAG + LLM Reranker for Resume ↔ JD Matching (ENHANCED UPLOAD VERSION)

🎯 FEATURES:
- 📁 UPLOAD .docx Resume (Colab button / Desktop picker / Paste fallback)
- 🔗 JD from URL or Paste
- 🧠 GPT-4o Reranking + Embeddings
- 📊 Detailed Score Breakdown
- ✅ Interactive & Production Ready

Usage: python resume_jd_matcher_fixed.py
"""

import os
import re
import json
import math
import time
import asyncio
import logging
from typing import List, Dict, Tuple, Optional, Set, Any
from collections import Counter
import numpy as np
import pandas as pd
from docx import Document
from rapidfuzz import process, fuzz
import spacy
from tqdm.auto import tqdm
import aiohttp
from bs4 import BeautifulSoup
import faiss
import html
from sklearn.feature_extraction.text import TfidfVectorizer
import getpass

# ---------- Logging ----------
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
logger = logging.getLogger(__name__)

# ---------- Load spaCy ----------
try:
    nlp = spacy.load("en_core_web_lg")
except Exception as e:
    logger.warning("spaCy model not found. Run: python -m spacy download en_core_web_lg")
    raise

# ---------- Configuration ----------
class Config:
    def __init__(self):
        self.embed_model = "text-embedding-3-large"
        self.llm_model = "gpt-4o"
        self.chunk_size = 150
        self.chunk_overlap = 30
        self.top_k_retrieve = 6
        self.weight_llm = 0.75
        self.weight_emb = 0.25
        self.max_new_tokens = 1200
        self.temperature = 0.1
        self.api_timeout = 120
        self.api_retries = 3
        self.concurrent_embeddings = 6
        self.criteria_keys = ["title_match", "level_seniority", "experience_years",
                             "technical_skills", "soft_skills", "qualifications",
                             "responsibilities_alignment", "domain_expertise"]
        self.redact_pii = False
        self.min_skill_length = 2
        self.max_skill_length = 50
        self.tfidf_top_n = 100

# ---------- 🚀 FIXED AUTOMATED UPLOAD FUNCTIONS ----------
import tempfile
import sys
import time
from pathlib import Path

def safe_filename(filename: str) -> str:
    """Clean filename for temp storage"""
    return re.sub(r'[<>:"/\\|?*]', '_', filename)

def prompt_for_resume_input() -> str:
    """🔥 FULLY AUTOMATED: Detects & handles Colab/Jupyter/Desktop in ONE GO"""
    print("\n📁 RESUME UPLOAD (AUTO-DETECTED)")

    # 1) COLAB - IMMEDIATE UPLOAD
    if "google.colab" in sys.modules:
        try:
            from google.colab import files
            print("✅ COLAB DETECTED - UPLOADING...")
            uploaded = files.upload()
            if uploaded:
                fname = next(iter(uploaded.keys()))
                dest = os.path.join(tempfile.gettempdir(), safe_filename(fname))
                with open(dest, "wb") as f:
                    f.write(uploaded[fname])
                print(f"✅ SAVED: {dest}")
                return dest
        except:
            pass

    # 2) JUPYTER - IMMEDIATE WIDGET
    if "ipykernel" in sys.modules:
        try:
            import ipywidgets as widgets
            from IPython.display import display
            print("✅ JUPYTER DETECTED - UPLOAD BELOW:")
            uploader = widgets.FileUpload(accept='.docx', multiple=False)
            display(uploader)
            # AUTO-WAIT 10s MAX
            for _ in range(20):  # 10 seconds
                if uploader.value:
                    uploaded_value = uploader.value
                    if isinstance(uploaded_value, dict):
                        name = list(uploaded_value.keys())[0]
                        content = uploaded_value[name]['content']
                    else:
                        item = uploaded_value[0]
                        name = item.get('name') or f"resume_{int(time.time())}.docx"
                        content = item.get('content')
                    dest = os.path.join(tempfile.gettempdir(), safe_filename(name))
                    with open(dest, "wb") as f:
                        f.write(content)
                    print(f"✅ JUPYTER SAVED: {dest}")
                    return dest
                time.sleep(0.5)
            print("⚠️ JUPYTER TIMEOUT - FALLBACK")
        except:
            pass

    # 3) DESKTOP - ONE-CLICK DIALOG
    try:
        import tkinter as tk
        from tkinter import filedialog
        root = tk.Tk()
        root.withdraw()
        print("✅ DESKTOP DETECTED - SELECT FILE")
        path = filedialog.askopenfilename(
            title="Select Resume (.docx)",
            filetypes=[("Word", "*.docx"), ("All", "*.*")]
        )
        if path and os.path.exists(path):
            print(f"✅ DESKTOP LOADED: {path}")
            return path
        print("⚠️ NO FILE - FALLBACK")
    except:
        pass

    # 4) ULTRA-FAST CLI (2 seconds total)
    print("⚡ QUICK: Enter path or press Enter:")
    path = input().strip()
    if path and os.path.exists(path):
        return path

    print("📝 PASTE RESUME (2 lines max):")
    text = input().strip()
    if text:
        return text

    raise RuntimeError("❌ No resume provided")

def prompt_for_jd_input() -> str:
    """JD input: URL or paste"""
    print("\n" + "="*50)
    print("📋 JOB DESCRIPTION INPUT:")
    print("="*50)
    print(" - URL: Paste LinkedIn/job URL (starts with http)")
    print(" - TEXT: Press Enter then paste JD")

    first = input("Enter JD URL or press Enter to paste: ").strip()
    if first:
        return first

    print("Paste JD text (end with blank line):")
    lines = []
    while True:
        try:
            line = input()
            if not line.strip():
                break
            lines.append(line)
        except EOFError:
            break
    text = "\n".join(lines).strip()
    if not text:
        raise RuntimeError("No JD provided")
    return text

def get_openai_api_key_interactive(provided_key: Optional[str] = None) -> str:
    """Get API key: provided → env → prompt"""
    if provided_key:
        return provided_key
    env_key = os.environ.get("OPENAI_API_KEY") or os.environ.get("OPENAI_APIKEY")
    if env_key:
        logger.info("Using OPENAI_API_KEY from environment")
        return env_key

    try:
        key = getpass.getpass("Enter OpenAI API key: ")
    except:
        key = input("Enter OpenAI API key: ").strip()

    if not key:
        raise RuntimeError("API key required")
    return key

# ---------- PII Redaction ----------
def redact_pii(text: str, nlp_model) -> str:
    if not text:
        return text
    doc = nlp_model(text)
    out = []
    last = 0
    for ent in doc.ents:
        if ent.label_ in {"PERSON", "GPE", "ORG"} or re.match(r".+@.+\..+", ent.text):
            out.append(text[last:ent.start_char])
            out.append(f"[REDACTED_{ent.label_}]")
            last = ent.end_char
    out.append(text[last:])
    return re.sub(r"\s{2,}", " ", "".join(out)).strip()

# ---------- Resume Parser ----------
class ResumeParser:
    SECTION_HEADERS = {
        "experience": r"(work\s+)?experience|employment\s+history|professional\s+background",
        "education": r"education|academic\s+background|qualifications",
        "skills": r"skills|technical\s+skills|competencies|expertise",
        "summary": r"summary|profile|objective|about\s+me",
        "projects": r"projects|portfolio",
        "certifications": r"certifications?|licenses?|credentials",
        "achievements": r"achievements?|awards?|honors?"
    }

    def __init__(self, text: str, nlp_model):
        self.text = text
        self.nlp = nlp_model
        self.sections = {}
        self._parse_sections()

    def _parse_sections(self):
        lines = self.text.split("\n")
        current_section = "header"
        current_content = []

        for line in lines:
            line_stripped = line.strip()
            if not line_stripped:
                continue

            matched_section = None
            for section_name, pattern in self.SECTION_HEADERS.items():
                if re.match(pattern, line_stripped, re.I) and len(line_stripped.split()) <= 5:
                    matched_section = section_name
                    break

            if matched_section:
                if current_content:
                    self.sections[current_section] = "\n".join(current_content)
                current_section = matched_section
                current_content = []
            else:
                current_content.append(line_stripped)

        if current_content:
            self.sections[current_section] = "\n".join(current_content)

        logger.info(f"Detected sections: {list(self.sections.keys())}")

    def get_section(self, section_name: str) -> str:
        return self.sections.get(section_name, "")

    def get_full_text(self) -> str:
        return self.text

# ---------- Dynamic Skills Extractor ----------
class DynamicSkillsExtractor:
    def __init__(self, nlp_model):
        self.nlp = nlp_model
        self.common_stop_words = {
            "experience", "work", "team", "project", "company", "role", "position",
            "responsible", "including", "using", "related", "various", "multiple",
            "required", "preferred", "ability", "knowledge", "understanding"
        }

    def extract_from_text(self, text: str, top_n: int = 100) -> Set[str]:
        skills = set()
        skills.update(self._extract_via_ner(text))
        skills.update(self._extract_via_tfidf(text, top_n))
        skills.update(self._extract_via_patterns(text))
        filtered = self._filter_skills(skills)
        logger.info(f"Extracted {len(filtered)} skills")
        return filtered

    def _extract_via_ner(self, text: str) -> Set[str]:
        doc = self.nlp(text)
        skills = set()
        for ent in doc.ents:
            if ent.label_ in {"PRODUCT", "ORG", "NORP", "LANGUAGE", "WORK_OF_ART"}:
                skills.add(ent.text.strip())
        for chunk in doc.noun_chunks:
            text_lower = chunk.text.lower()
            if any(indicator in text_lower for indicator in
                   ["software", "tool", "framework", "language", "platform", "system"]):
                skills.add(chunk.text.strip())
        return skills

    def _extract_via_tfidf(self, text: str, top_n: int) -> Set[str]:
        try:
            vectorizer = TfidfVectorizer(
                max_features=top_n, ngram_range=(1, 3), stop_words='english',
                token_pattern=r'\b[A-Za-z][A-Za-z0-9\.\+\#\-]*\b'
            )
            tfidf_matrix = vectorizer.fit_transform([text])
            feature_names = vectorizer.get_feature_names_out()
            scores = tfidf_matrix.toarray()[0]
            top_indices = scores.argsort()[-top_n:][::-1]
            return {feature_names[i] for i in top_indices if scores[i] > 0.01}
        except:
            return set()

    def _extract_via_patterns(self, text: str) -> Set[str]:
        skills = set()
        skills.update(re.findall(r'\b([A-Z][a-z]+(?:[A-Z][a-z]+)*(?:\.[a-z]+)?)\b', text))
        skills.update(re.findall(r'\b([A-Z]{2,6})\b', text))
        potential = re.findall(r'\b([A-Za-z]+(?:\s+\d+(?:\.\d+)?)?)\b', text)
        skills.update([p for p in potential if any(c.isdigit() for c in p)])
        skills.update(re.findall(r'\b([A-Za-z]+-[A-Za-z]+(?:-[A-Za-z]+)?)\b', text))
        return skills

    def _filter_skills(self, skills: Set[str]) -> Set[str]:
        filtered = set()
        for skill in skills:
            skill_clean = skill.strip()
            skill_lower = skill_clean.lower()
            if (2 <= len(skill_clean) <= 50 and
                skill_lower not in self.common_stop_words and
                re.search(r'[A-Za-z]', skill_clean) and
                skill_lower not in {"the", "and", "for", "with", "from", "this", "that", "will", "can"}):
                filtered.add(skill_clean)
        return filtered

# ---------- JD Extraction ----------
async def load_jd_from_url(url: str, session: aiohttp.ClientSession, timeout: int = 30) -> str:
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    }
    async with session.get(url, headers=headers, timeout=timeout, allow_redirects=True) as r:
        r.raise_for_status()
        html_text = await r.text()

    soup = BeautifulSoup(html_text, "html.parser")
    linkedin_selectors = [
        {"name": "div", "attrs": {"class": "description__text"}},
        {"name": "div", "attrs": {"class": "show-more-less-html__markup"}},
        {"name": "section", "attrs": {"class": "description"}},
    ]
    generic_selectors = [
        {"name": "div", "attrs": {"class": re.compile(r"(?i)(job[-_]?description|description)")}},
        {"name": "section", "attrs": {"class": re.compile(r"(?i)(job|description)")}},
    ]

    all_selectors = linkedin_selectors + generic_selectors
    text = ""
    for sel in all_selectors:
        node = soup.find(sel.get("name"), sel.get("attrs"))
        if node:
            text = node.get_text(separator="\n", strip=True)
            if len(text.split()) >= 50:
                break

    if not text or len(text.split()) < 50:
        paragraphs = soup.find_all("p")
        text = "\n".join([p.get_text(strip=True) for p in paragraphs])

    return re.sub(r"\n{3,}", "\n\n", html.unescape(text)).strip()

def read_docx(path: str) -> str:
    doc = Document(path)
    paragraphs = []
    for p in doc.paragraphs:
        text = p.text.strip()
        if text:
            paragraphs.append(text)
    for table in doc.tables:
        for row in table.rows:
            for cell in row.cells:
                text = cell.text.strip()
                if text:
                    paragraphs.append(text)
    return "\n".join(paragraphs)

def extract_experience_years(text: str) -> Optional[int]:
    patterns = [
        r"(\d+)\+?\s*(?:years?|yrs?)[\s\w]*(?:of\s+)?(?:experience|exp)",
        r"(?:minimum|at least)\s+(\d+)\s*(?:years?|yrs?)",
        r"(\d+)(?:\s*-\s*(\d+))?\s*(?:years?|yrs?)",
    ]
    max_years = None
    for pattern in patterns:
        matches = re.findall(pattern, text, re.I)
        for match in matches:
            nums = [int(n) for n in match if n and n.isdigit()]
            if nums:
                year = max(nums)
                if max_years is None or year > max_years:
                    max_years = year
    return max_years

# ---------- Main Matcher ----------
class ResumeJDMatcher:
    def __init__(self, config: Config, openai_api_key: str):
        self.config = config
        self.openai_api_key = openai_api_key  # ✅ BUGFIX: actually stores key
        self.resume_text: Optional[str] = None
        self.resume_parser: Optional[ResumeParser] = None
        self.jd_text: Optional[str] = None
        self.jd_criteria: Optional[Dict] = None
        self.jd_skills: Set[str] = set()
        self.resume_skills: Set[str] = set()
        self.resume_chunks: List[str] = []
        self.index: Optional[faiss.Index] = None
        self.resume_embs: Optional[np.ndarray] = None
        self._aiohttp_session: Optional[aiohttp.ClientSession] = None
        self._embed_semaphore = asyncio.Semaphore(self.config.concurrent_embeddings)
        self.skills_extractor = DynamicSkillsExtractor(nlp)

    async def ensure_session(self) -> aiohttp.ClientSession:
        if self._aiohttp_session is None:
            self._aiohttp_session = aiohttp.ClientSession()
        return self._aiohttp_session

    async def close(self):
        if self._aiohttp_session:
            await self._aiohttp_session.close()

    async def load_resume(self, path_or_text: str) -> None:
        if path_or_text.endswith(".docx") and os.path.exists(path_or_text):
            self.resume_text = read_docx(path_or_text)
            logger.info(f"✅ Loaded .docx: {path_or_text}")
        else:
            self.resume_text = path_or_text

        if not self.resume_text.strip():
            raise ValueError("Empty resume")

        self.resume_parser = ResumeParser(self.resume_text, nlp)
        self.resume_skills = self.skills_extractor.extract_from_text(self.resume_text)
        logger.info(f"Resume: {len(self.resume_text.split())} words, {len(self.resume_skills)} skills")

    async def load_jd(self, url_or_text: str) -> None:
        session = await self.ensure_session()
        if url_or_text.startswith("http"):
            self.jd_text = await load_jd_from_url(url_or_text, session)
            logger.info(f"✅ JD from URL: {len(self.jd_text.split())} words")
        else:
            self.jd_text = url_or_text

        if not self.jd_text.strip():
            raise ValueError("Empty JD")

        self.jd_skills = self.skills_extractor.extract_from_text(self.jd_text)
        logger.info(f"JD: {len(self.jd_text.split())} words, {len(self.jd_skills)} skills")

    def extract_jd_criteria(self) -> None:
        doc = nlp(self.jd_text)
        title = ""
        for line in self.jd_text.split("\n")[:15]:
            line_clean = line.strip()
            if 3 <= len(line_clean.split()) <= 12 and not line_clean.endswith(":"):
                title = line_clean
                break

        level_keywords = ["senior", "junior", "lead", "principal", "staff", "manager"]
        level = next((token.text for sent in list(doc.sents)[:20]
                     for token in sent if token.lower_ in level_keywords), "")

        exp_years = extract_experience_years(self.jd_text)
        qual_keywords = ["bachelor", "master", "phd", "degree", "certification"]
        quals = [sent.text.strip() for sent in doc.sents
                if any(kw in sent.text.lower() for kw in qual_keywords)][:10]

        action_verbs = {"develop", "design", "lead", "manage", "implement", "build"}
        resps = [sent.text.strip() for sent in doc.sents
                if any(token.lemma_ in action_verbs for token in sent)][:15]

        domain_keywords = ["healthcare", "finance", "fintech", "ecommerce", "saas",
                          "cloud", "ai", "machine learning", "data science"]
        domain = [kw for kw in domain_keywords if kw in self.jd_text.lower()]

        self.jd_criteria = {
            "title": title, "level": level, "experience_years": exp_years,
            "skills": sorted(list(self.jd_skills))[:50],
            "qualifications": quals, "responsibilities": resps, "domain": domain
        }

    def chunk_resume(self) -> None:
        doc = nlp(self.resume_text)
        chunks = []
        current_chunk = []
        current_words = 0

        for sent in doc.sents:
            sent_text = sent.text.strip()
            sent_words = len(sent_text.split())
            if current_words + sent_words > self.config.chunk_size and current_chunk:
                chunk_text = " ".join(current_chunk)
                chunks.append(chunk_text)
                current_chunk = []
                current_words = 0
            current_chunk.append(sent_text)
            current_words += sent_words

        if current_chunk:
            chunks.append(" ".join(current_chunk))

        self.resume_chunks = chunks
        logger.info(f"Chunked into {len(chunks)} segments")

    async def _openai_embedding(self, text: str) -> np.ndarray:
        session = await self.ensure_session()
        endpoint = "https://api.openai.com/v1/embeddings"
        headers = {"Authorization": f"Bearer {self.openai_api_key}", "Content-Type": "application/json"}
        payload = {"input": text[:8000], "model": self.config.embed_model}

        async with session.post(endpoint, headers=headers, json=payload,
                               timeout=self.config.api_timeout) as r:
            data = await r.json()
        vec = np.array(data["data"][0]["embedding"], dtype=np.float32)
        norm = np.linalg.norm(vec)
        if norm > 0:
            vec /= norm
        return vec

    async def embed_resume(self) -> None:
        if not self.resume_chunks:
            self.chunk_resume()

        tasks = [self._openai_embedding(chunk) for chunk in self.resume_chunks]
        embeddings = []
        for task in tqdm(asyncio.as_completed(tasks), total=len(tasks), desc="Embedding"):
            embeddings.append(await task)

        self.resume_embs = np.vstack(embeddings).astype("float32")
        dim = self.resume_embs.shape[1]
        self.index = faiss.IndexFlatIP(dim)
        self.index.add(self.resume_embs)
        logger.info(f"FAISS: {self.resume_embs.shape[0]} vectors")

    async def retrieve_top_k(self, query_text: str, k: int) -> List[Dict]:
        query_vec = await self._openai_embedding(query_text)
        norm = np.linalg.norm(query_vec)
        if norm > 0:
            query_vec /= norm
        sims, indices = self.index.search(np.expand_dims(query_vec, 0), k)
        return [{"idx": int(idx), "similarity": float(sim), "text": self.resume_chunks[idx]}
                for sim, idx in zip(sims[0], indices[0]) if idx >= 0]

    def create_criterion_query(self, criterion_key: str) -> str:
        criteria = self.jd_criteria or {}
        queries = {
            "title_match": f"Job title: {criteria.get('title', 'Unknown')}. Relevant roles.",
            "level_seniority": f"Seniority: {criteria.get('level', 'Not specified')}. Leadership.",
            "experience_years": f"Required: {criteria.get('experience_years', 0)} years experience.",
            "technical_skills": f"Skills: {', '.join(list(self.jd_skills)[:20])}",
            "soft_skills": "Communication, leadership, teamwork, problem-solving.",
            "qualifications": f"Education: {'; '.join(criteria.get('qualifications', [])[:3])}",
            "responsibilities_alignment": f"Responsibilities: {' '.join(criteria.get('responsibilities', [])[:5])}",
            "domain_expertise": f"Domain: {', '.join(criteria.get('domain', ['general']))}"
        }
        return queries.get(criterion_key, self.jd_text[:500])

    def _get_enhanced_cot_prompt(self, criterion: str, query: str, passages: str) -> str:
        skill_overlap = len(self.jd_skills & self.resume_skills)
        skill_overlap_pct = (skill_overlap / len(self.jd_skills) * 100) if self.jd_skills else 0

        return f"""EXPERT RECRUITER EVALUATION

CRITERION: {criterion}
QUERY: {query}

RESUME EVIDENCE:
{passages}


EVALUATE:
1. Extract evidence (direct/transferable/gaps). Be strict: only direct matches get high scores; transferable skills score medium; mismatches or gaps score low.
2. Assess relevance and recency
3. Synthesize holistically

OUTPUT:
PASSAGE_ANALYSIS: [for each passage]
- Evidence: [specific facts]
- Reasoning: [why relevant]
PASSAGE_RELEVANCE_SCORE: [0-100]


CONTEXT:
- JD Skills: {len(self.jd_skills)} | Resume Skills: {len(self.resume_skills)}
- Overlap: {skill_overlap} ({skill_overlap_pct:.1f}%)
- Exp Required: {self.jd_criteria.get('experience_years', 'N/A')} years

**SCORE THIS CRITERION (0-100):**
90-100: Exceeds ✓ (perfect direct match, exceeds requirements)
75-89: Strong ✓ (strong direct match, minor gaps)
60-74: Adequate ✓ (acceptable with some transferable skills)
40-59: Partial ⚠ (partial match, significant gaps)
20-39: Weak ⚠ (weak relevance, major mismatches)
0-19: No fit ✗ (no relevance or complete mismatch)

OUTPUT EXACTLY:
FINAL_CRITERION_SCORE: [0-100]

RECOMMENDATIONS:
- [3-5 bullet points on improvements or strengths, tailored to this criterion. If high score, highlight strengths; if low, suggest specific ways to improve resume or skills for better fit.]

JD REFERENCE:
{self.jd_text[:1500]}"""

# ---------- LLM Helpers ----------
async def _openai_chat_with_retry(session: aiohttp.ClientSession, api_key: str, messages: list,
                                  model: str, timeout: int = 120, retries: int = 3,
                                  temperature: float = 0.1, max_tokens: int = 1024) -> dict:
    endpoint = "https://api.openai.com/v1/chat/completions"
    headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}
    payload = {"model": model, "messages": messages, "temperature": temperature, "max_tokens": max_tokens}

    for attempt in range(retries):
        try:
            async with session.post(endpoint, headers=headers, json=payload, timeout=timeout) as r:
                if r.status != 200:
                    raise RuntimeError(f"API {r.status}")
                return await r.json()
        except Exception as e:
            if attempt == retries - 1:
                raise
            await asyncio.sleep(2 ** attempt)

def parse_final_score_from_text(text: str) -> Optional[int]:
    m = re.search(r"FINAL_CRITERION_SCORE\s*:\s*(\d{1,3})", text)
    if m:
        return max(0, min(100, int(m.group(1))))
    m2 = re.search(r"\b([0-9]{1,3})\b", text)
    if m2:
        s = int(m2.group(1))
        if 0 <= s <= 100:
            return s
    return None

# ---------- 🎯 MAIN FLOW ----------
async def main_flow(resume_path_or_text: Optional[str] = None,
                    jd_url_or_text: Optional[str] = None,
                    config: Optional[Config] = None,
                    openai_api_key: Optional[str] = None,
                    show_verbose: bool = True) -> Dict[str, Any]:

    if config is None:
        config = Config()
    openai_api_key = get_openai_api_key_interactive(openai_api_key)
    matcher = ResumeJDMatcher(config, openai_api_key)

    try:
        session = await matcher.ensure_session()

        # 🚀 INTERACTIVE INPUTS
        if resume_path_or_text is None:
            resume_path_or_text = prompt_for_resume_input()
        if jd_url_or_text is None:
            jd_url_or_text = prompt_for_jd_input()

        await matcher.load_resume(resume_path_or_text)
        await matcher.load_jd(jd_url_or_text)
        matcher.extract_jd_criteria()
        await matcher.embed_resume()

        criterion_results = {}
        for crit in matcher.config.criteria_keys:
            query = matcher.create_criterion_query(crit)
            passages = await matcher.retrieve_top_k(query, matcher.config.top_k_retrieve)

            passages_text = "\n\n".join([f"Passage {i+1}: {p['text']}"
                                       for i, p in enumerate(passages)])
            cot_prompt = matcher._get_enhanced_cot_prompt(crit, query, passages_text)

            messages = [{"role": "system", "content": "Expert recruiter."},
                       {"role": "user", "content": cot_prompt}]

            chat_resp = await _openai_chat_with_retry(session, openai_api_key, messages,
                                                    matcher.config.llm_model,
                                                    max_tokens=matcher.config.max_new_tokens)
            llm_text = chat_resp["choices"][0]["message"]["content"]
            llm_score = parse_final_score_from_text(llm_text)

            avg_sim = np.mean([p["similarity"] for p in passages]) * 100 if passages else 0
            emb_score = float(avg_sim)
            combined_score = (config.weight_llm * (llm_score or emb_score) +
                            config.weight_emb * emb_score)
            combined_score = max(0, min(100, combined_score))

            criterion_results[crit] = {
                "llm_score": llm_score, "emb_score": round(emb_score, 2),
                "combined_score": round(combined_score, 2),
                "explanation": llm_text
            }

            if show_verbose:
                logger.info(f"{crit}: {combined_score:.1f} (LLM:{llm_score}, Emb:{emb_score:.1f})")

        overall_score = np.mean([v["combined_score"] for v in criterion_results.values()])

        result = {
            "overall_score": round(overall_score, 2),
            "criteria": criterion_results,
            "jd_criteria": matcher.jd_criteria,
            "resume_skills": sorted(list(matcher.resume_skills))[:50],
            "jd_skills": sorted(list(matcher.jd_skills))[:50]
        }

        print(f"\n🎉 FINAL SCORE: {result['overall_score']:.1f}/100")
        for k, v in result["criteria"].items():
            print(f"  {k}: LLM {v['llm_score']}, EMB {v['emb_score']:.1f}, Average {v['combined_score']:.1f}")
            print(f"    Explanation and Recommendations:\n{v['explanation']}\n")

        return result

    finally:
        await matcher.close()

# ---------- RUN (portable: script + notebook friendly) ----------
if __name__ == "__main__":
    try:
        # Try the simple, normal case first
        try:
            asyncio.run(main_flow())
        except RuntimeError as re:
            # Happens when running inside an existing event loop (e.g. Jupyter/Colab)
            logger.warning("Detected running event loop; attempting notebook-friendly execution.")
            try:
                # Prefer using nest_asyncio if available
                import nest_asyncio
                nest_asyncio.apply()
                loop = asyncio.get_event_loop()
                loop.run_until_complete(main_flow())
            except Exception:
                # If nest_asyncio is not available or something else failed,
                # provide clear instruction for notebook execution.
                logger.error("Could not auto-run inside the existing event loop.")
                logger.info("If you're in a Jupyter/Colab notebook, run: `await main_flow()`")
                raise
    except Exception as e:
        logger.exception(f"Failed: {e}")
```

```python
"""
Advanced Hybrid RAG + LLM Reranker for Resume ↔ JD Matching (ENHANCED UPLOAD VERSION)

🎯 FEATURES:
- 📁 UPLOAD .docx Resume (Colab button / Desktop picker / Paste fallback)
- 🔗 JD from URL or Paste
- 🧠 GPT-4o Reranking + Embeddings
- 📊 Detailed Score Breakdown
- ✅ Interactive & Production Ready

Usage: python resume_jd_matcher_fixed.py
"""

import os
import re
import json
import math
import time
import asyncio
import logging
from typing import List, Dict, Tuple, Optional, Set, Any
from collections import Counter
import numpy as np
import pandas as pd
from docx import Document
from rapidfuzz import process, fuzz
import spacy
from tqdm.auto import tqdm
import aiohttp
from bs4 import BeautifulSoup
import faiss
import html
from sklearn.feature_extraction.text import TfidfVectorizer
import getpass

# ---------- Logging ----------
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
logger = logging.getLogger(__name__)

# ---------- Load spaCy ----------
try:
    nlp = spacy.load("en_core_web_lg")
except Exception as e:
    logger.warning("spaCy model not found. Run: python -m spacy download en_core_web_lg")
    raise

# ---------- Configuration ----------
class Config:
    def __init__(self):
        self.embed_model = "text-embedding-3-large"
        self.llm_model = "gpt-4o"
        self.chunk_size = 150
        self.chunk_overlap = 30
        self.top_k_retrieve = 6
        self.weight_llm = 0.75
        self.weight_emb = 0.25
        self.max_new_tokens = 1200
        self.temperature = 0.1
        self.api_timeout = 120
        self.api_retries = 3
        self.concurrent_embeddings = 6
        self.criteria_keys = ["title_match", "level_seniority", "experience_years",
                             "technical_skills", "soft_skills", "qualifications",
                             "responsibilities_alignment", "domain_expertise"]
        self.redact_pii = False
        self.min_skill_length = 2
        self.max_skill_length = 50
        self.tfidf_top_n = 100

# ---------- 🚀 FIXED AUTOMATED UPLOAD FUNCTIONS ----------
import tempfile
import sys
import time
from pathlib import Path

def safe_filename(filename: str) -> str:
    """Clean filename for temp storage"""
    return re.sub(r'[<>:"/\\|?*]', '_', filename)

def prompt_for_resume_input() -> str:
    """🔥 FULLY AUTOMATED: Detects & handles Colab/Jupyter/Desktop in ONE GO"""
    print("\n📁 RESUME UPLOAD (AUTO-DETECTED)")

    # 1) COLAB - IMMEDIATE UPLOAD
    if "google.colab" in sys.modules:
        try:
            from google.colab import files
            print("✅ COLAB DETECTED - UPLOADING...")
            uploaded = files.upload()
            if uploaded:
                fname = next(iter(uploaded.keys()))
                dest = os.path.join(tempfile.gettempdir(), safe_filename(fname))
                with open(dest, "wb") as f:
                    f.write(uploaded[fname])
                print(f"✅ SAVED: {dest}")
                return dest
        except:
            pass

    # 2) JUPYTER - IMMEDIATE WIDGET
    if "ipykernel" in sys.modules:
        try:
            import ipywidgets as widgets
            from IPython.display import display
            print("✅ JUPYTER DETECTED - UPLOAD BELOW:")
            uploader = widgets.FileUpload(accept='.docx', multiple=False)
            display(uploader)
            # AUTO-WAIT 10s MAX
            for _ in range(20):  # 10 seconds
                if uploader.value:
                    uploaded_value = uploader.value
                    if isinstance(uploaded_value, dict):
                        name = list(uploaded_value.keys())[0]
                        content = uploaded_value[name]['content']
                    else:
                        item = uploaded_value[0]
                        name = item.get('name') or f"resume_{int(time.time())}.docx"
                        content = item.get('content')
                    dest = os.path.join(tempfile.gettempdir(), safe_filename(name))
                    with open(dest, "wb") as f:
                        f.write(content)
                    print(f"✅ JUPYTER SAVED: {dest}")
                    return dest
                time.sleep(0.5)
            print("⚠️ JUPYTER TIMEOUT - FALLBACK")
        except:
            pass

    # 3) DESKTOP - ONE-CLICK DIALOG
    try:
        import tkinter as tk
        from tkinter import filedialog
        root = tk.Tk()
        root.withdraw()
        print("✅ DESKTOP DETECTED - SELECT FILE")
        path = filedialog.askopenfilename(
            title="Select Resume (.docx)",
            filetypes=[("Word", "*.docx"), ("All", "*.*")]
        )
        if path and os.path.exists(path):
            print(f"✅ DESKTOP LOADED: {path}")
            return path
        print("⚠️ NO FILE - FALLBACK")
    except:
        pass

    # 4) ULTRA-FAST CLI (2 seconds total)
    print("⚡ QUICK: Enter path or press Enter:")
    path = input().strip()
    if path and os.path.exists(path):
        return path

    print("📝 PASTE RESUME (2 lines max):")
    text = input().strip()
    if text:
        return text

    raise RuntimeError("❌ No resume provided")

def prompt_for_jd_input() -> str:
    """JD input: URL or paste"""
    print("\n" + "="*50)
    print("📋 JOB DESCRIPTION INPUT:")
    print("="*50)
    print(" - URL: Paste LinkedIn/job URL (starts with http)")
    print(" - TEXT: Press Enter then paste JD")

    first = input("Enter JD URL or press Enter to paste: ").strip()
    if first:
        return first

    print("Paste JD text (end with blank line):")
    lines = []
    while True:
        try:
            line = input()
            if not line.strip():
                break
            lines.append(line)
        except EOFError:
            break
    text = "\n".join(lines).strip()
    if not text:
        raise RuntimeError("No JD provided")
    return text

def get_openai_api_key_interactive(provided_key: Optional[str] = None) -> str:
    """Get API key: provided → env → prompt"""
    if provided_key:
        return provided_key
    env_key = os.environ.get("OPENAI_API_KEY") or os.environ.get("OPENAI_APIKEY")
    if env_key:
        logger.info("Using OPENAI_API_KEY from environment")
        return env_key

    try:
        key = getpass.getpass("Enter OpenAI API key: ")
    except:
        key = input("Enter OpenAI API key: ").strip()

    if not key:
        raise RuntimeError("API key required")
    return key

# ---------- PII Redaction ----------
def redact_pii(text: str, nlp_model) -> str:
    if not text:
        return text
    doc = nlp_model(text)
    out = []
    last = 0
    for ent in doc.ents:
        if ent.label_ in {"PERSON", "GPE", "ORG"} or re.match(r".+@.+\..+", ent.text):
            out.append(text[last:ent.start_char])
            out.append(f"[REDACTED_{ent.label_}]")
            last = ent.end_char
    out.append(text[last:])
    return re.sub(r"\s{2,}", " ", "".join(out)).strip()

# ---------- Resume Parser ----------
class ResumeParser:
    SECTION_HEADERS = {
        "experience": r"(work\s+)?experience|employment\s+history|professional\s+background",
        "education": r"education|academic\s+background|qualifications",
        "skills": r"skills|technical\s+skills|competencies|expertise",
        "summary": r"summary|profile|objective|about\s+me",
        "projects": r"projects|portfolio",
        "certifications": r"certifications?|licenses?|credentials",
        "achievements": r"achievements?|awards?|honors?"
    }

    def __init__(self, text: str, nlp_model):
        self.text = text
        self.nlp = nlp_model
        self.sections = {}
        self._parse_sections()

    def _parse_sections(self):
        lines = self.text.split("\n")
        current_section = "header"
        current_content = []

        for line in lines:
            line_stripped = line.strip()
            if not line_stripped:
                continue

            matched_section = None
            for section_name, pattern in self.SECTION_HEADERS.items():
                if re.match(pattern, line_stripped, re.I) and len(line_stripped.split()) <= 5:
                    matched_section = section_name
                    break

            if matched_section:
                if current_content:
                    self.sections[current_section] = "\n".join(current_content)
                current_section = matched_section
                current_content = []
            else:
                current_content.append(line_stripped)

        if current_content:
            self.sections[current_section] = "\n".join(current_content)

        logger.info(f"Detected sections: {list(self.sections.keys())}")

    def get_section(self, section_name: str) -> str:
        return self.sections.get(section_name, "")

    def get_full_text(self) -> str:
        return self.text

# ---------- Dynamic Skills Extractor ----------
class DynamicSkillsExtractor:
    def __init__(self, nlp_model):
        self.nlp = nlp_model
        self.common_stop_words = {
            "experience", "work", "team", "project", "company", "role", "position",
            "responsible", "including", "using", "related", "various", "multiple",
            "required", "preferred", "ability", "knowledge", "understanding"
        }

    def extract_from_text(self, text: str, top_n: int = 100) -> Set[str]:
        skills = set()
        skills.update(self._extract_via_ner(text))
        skills.update(self._extract_via_tfidf(text, top_n))
        skills.update(self._extract_via_patterns(text))
        filtered = self._filter_skills(skills)
        logger.info(f"Extracted {len(filtered)} skills")
        return filtered

    def _extract_via_ner(self, text: str) -> Set[str]:
        doc = self.nlp(text)
        skills = set()
        for ent in doc.ents:
            if ent.label_ in {"PRODUCT", "ORG", "NORP", "LANGUAGE", "WORK_OF_ART"}:
                skills.add(ent.text.strip())
        for chunk in doc.noun_chunks:
            text_lower = chunk.text.lower()
            if any(indicator in text_lower for indicator in
                   ["software", "tool", "framework", "language", "platform", "system"]):
                skills.add(chunk.text.strip())
        return skills

    def _extract_via_tfidf(self, text: str, top_n: int) -> Set[str]:
        try:
            vectorizer = TfidfVectorizer(
                max_features=top_n, ngram_range=(1, 3), stop_words='english',
                token_pattern=r'\b[A-Za-z][A-Za-z0-9\.\+\#\-]*\b'
            )
            tfidf_matrix = vectorizer.fit_transform([text])
            feature_names = vectorizer.get_feature_names_out()
            scores = tfidf_matrix.toarray()[0]
            top_indices = scores.argsort()[-top_n:][::-1]
            return {feature_names[i] for i in top_indices if scores[i] > 0.01}
        except:
            return set()

    def _extract_via_patterns(self, text: str) -> Set[str]:
        skills = set()
        skills.update(re.findall(r'\b([A-Z][a-z]+(?:[A-Z][a-z]+)*(?:\.[a-z]+)?)\b', text))
        skills.update(re.findall(r'\b([A-Z]{2,6})\b', text))
        potential = re.findall(r'\b([A-Za-z]+(?:\s+\d+(?:\.\d+)?)?)\b', text)
        skills.update([p for p in potential if any(c.isdigit() for c in p)])
        skills.update(re.findall(r'\b([A-Za-z]+-[A-Za-z]+(?:-[A-Za-z]+)?)\b', text))
        return skills

    def _filter_skills(self, skills: Set[str]) -> Set[str]:
        filtered = set()
        for skill in skills:
            skill_clean = skill.strip()
            skill_lower = skill_clean.lower()
            if (2 <= len(skill_clean) <= 50 and
                skill_lower not in self.common_stop_words and
                re.search(r'[A-Za-z]', skill_clean) and
                skill_lower not in {"the", "and", "for", "with", "from", "this", "that", "will", "can"}):
                filtered.add(skill_clean)
        return filtered

# ---------- JD Extraction ----------
async def load_jd_from_url(url: str, session: aiohttp.ClientSession, timeout: int = 30) -> str:
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    }
    async with session.get(url, headers=headers, timeout=timeout, allow_redirects=True) as r:
        r.raise_for_status()
        html_text = await r.text()

    soup = BeautifulSoup(html_text, "html.parser")
    linkedin_selectors = [
        {"name": "div", "attrs": {"class": "description__text"}},
        {"name": "div", "attrs": {"class": "show-more-less-html__markup"}},
        {"name": "section", "attrs": {"class": "description"}},
    ]
    generic_selectors = [
        {"name": "div", "attrs": {"class": re.compile(r"(?i)(job[-_]?description|description)")}},
        {"name": "section", "attrs": {"class": re.compile(r"(?i)(job|description)")}},
    ]

    all_selectors = linkedin_selectors + generic_selectors
    text = ""
    for sel in all_selectors:
        node = soup.find(sel.get("name"), sel.get("attrs"))
        if node:
            text = node.get_text(separator="\n", strip=True)
            if len(text.split()) >= 50:
                break

    if not text or len(text.split()) < 50:
        paragraphs = soup.find_all("p")
        text = "\n".join([p.get_text(strip=True) for p in paragraphs])

    return re.sub(r"\n{3,}", "\n\n", html.unescape(text)).strip()

def read_docx(path: str) -> str:
    doc = Document(path)
    paragraphs = []
    for p in doc.paragraphs:
        text = p.text.strip()
        if text:
            paragraphs.append(text)
    for table in doc.tables:
        for row in table.rows:
            for cell in row.cells:
                text = cell.text.strip()
                if text:
                    paragraphs.append(text)
    return "\n".join(paragraphs)

def extract_experience_years(text: str) -> Optional[int]:
    patterns = [
        r"(\d+)\+?\s*(?:years?|yrs?)[\s\w]*(?:of\s+)?(?:experience|exp)",
        r"(?:minimum|at least)\s+(\d+)\s*(?:years?|yrs?)",
        r"(\d+)(?:\s*-\s*(\d+))?\s*(?:years?|yrs?)",
    ]
    max_years = None
    for pattern in patterns:
        matches = re.findall(pattern, text, re.I)
        for match in matches:
            nums = [int(n) for n in match if n and n.isdigit()]
            if nums:
                year = max(nums)
                if max_years is None or year > max_years:
                    max_years = year
    return max_years

# ---------- Main Matcher ----------
class ResumeJDMatcher:
    def __init__(self, config: Config, openai_api_key: str):
        self.config = config
        self.openai_api_key = openai_api_key  # ✅ BUGFIX: actually stores key
        self.resume_text: Optional[str] = None
        self.resume_parser: Optional[ResumeParser] = None
        self.jd_text: Optional[str] = None
        self.jd_criteria: Optional[Dict] = None
        self.jd_skills: Set[str] = set()
        self.resume_skills: Set[str] = set()
        self.resume_chunks: List[str] = []
        self.index: Optional[faiss.Index] = None
        self.resume_embs: Optional[np.ndarray] = None
        self._aiohttp_session: Optional[aiohttp.ClientSession] = None
        self._embed_semaphore = asyncio.Semaphore(self.config.concurrent_embeddings)
        self.skills_extractor = DynamicSkillsExtractor(nlp)

    async def ensure_session(self) -> aiohttp.ClientSession:
        if self._aiohttp_session is None:
            self._aiohttp_session = aiohttp.ClientSession()
        return self._aiohttp_session

    async def close(self):
        if self._aiohttp_session:
            await self._aiohttp_session.close()

    async def load_resume(self, path_or_text: str) -> None:
        if path_or_text.endswith(".docx") and os.path.exists(path_or_text):
            self.resume_text = read_docx(path_or_text)
            logger.info(f"✅ Loaded .docx: {path_or_text}")
        else:
            self.resume_text = path_or_text

        if not self.resume_text.strip():
            raise ValueError("Empty resume")

        self.resume_parser = ResumeParser(self.resume_text, nlp)
        self.resume_skills = self.skills_extractor.extract_from_text(self.resume_text)
        logger.info(f"Resume: {len(self.resume_text.split())} words, {len(self.resume_skills)} skills")

    async def load_jd(self, url_or_text: str) -> None:
        session = await self.ensure_session()
        if url_or_text.startswith("http"):
            self.jd_text = await load_jd_from_url(url_or_text, session)
            logger.info(f"✅ JD from URL: {len(self.jd_text.split())} words")
        else:
            self.jd_text = url_or_text

        if not self.jd_text.strip():
            raise ValueError("Empty JD")

        self.jd_skills = self.skills_extractor.extract_from_text(self.jd_text)
        logger.info(f"JD: {len(self.jd_text.split())} words, {len(self.jd_skills)} skills")

    def extract_jd_criteria(self) -> None:
        doc = nlp(self.jd_text)
        title = ""
        for line in self.jd_text.split("\n")[:15]:
            line_clean = line.strip()
            if 3 <= len(line_clean.split()) <= 12 and not line_clean.endswith(":"):
                title = line_clean
                break

        level_keywords = ["senior", "junior", "lead", "principal", "staff", "manager"]
        level = next((token.text for sent in list(doc.sents)[:20]
                     for token in sent if token.lower_ in level_keywords), "")

        exp_years = extract_experience_years(self.jd_text)
        qual_keywords = ["bachelor", "master", "phd", "degree", "certification"]
        quals = [sent.text.strip() for sent in doc.sents
                if any(kw in sent.text.lower() for kw in qual_keywords)][:10]

        action_verbs = {"develop", "design", "lead", "manage", "implement", "build"}
        resps = [sent.text.strip() for sent in doc.sents
                if any(token.lemma_ in action_verbs for token in sent)][:15]

        domain_keywords = ["healthcare", "finance", "fintech", "ecommerce", "saas",
                          "cloud", "ai", "machine learning", "data science"]
        domain = [kw for kw in domain_keywords if kw in self.jd_text.lower()]

        self.jd_criteria = {
            "title": title, "level": level, "experience_years": exp_years,
            "skills": sorted(list(self.jd_skills))[:50],
            "qualifications": quals, "responsibilities": resps, "domain": domain
        }

    def chunk_resume(self) -> None:
        doc = nlp(self.resume_text)
        chunks = []
        current_chunk = []
        current_words = 0

        for sent in doc.sents:
            sent_text = sent.text.strip()
            sent_words = len(sent_text.split())
            if current_words + sent_words > self.config.chunk_size and current_chunk:
                chunk_text = " ".join(current_chunk)
                chunks.append(chunk_text)
                current_chunk = []
                current_words = 0
            current_chunk.append(sent_text)
            current_words += sent_words

        if current_chunk:
            chunks.append(" ".join(current_chunk))

        self.resume_chunks = chunks
        logger.info(f"Chunked into {len(chunks)} segments")

    async def _openai_embedding(self, text: str) -> np.ndarray:
        session = await self.ensure_session()
        endpoint = "https://api.openai.com/v1/embeddings"
        headers = {"Authorization": f"Bearer {self.openai_api_key}", "Content-Type": "application/json"}
        payload = {"input": text[:8000], "model": self.config.embed_model}

        async with session.post(endpoint, headers=headers, json=payload,
                               timeout=self.config.api_timeout) as r:
            data = await r.json()
        vec = np.array(data["data"][0]["embedding"], dtype=np.float32)
        norm = np.linalg.norm(vec)
        if norm > 0:
            vec /= norm
        return vec

    async def embed_resume(self) -> None:
        if not self.resume_chunks:
            self.chunk_resume()

        tasks = [self._openai_embedding(chunk) for chunk in self.resume_chunks]
        embeddings = []
        for task in tqdm(asyncio.as_completed(tasks), total=len(tasks), desc="Embedding"):
            embeddings.append(await task)

        self.resume_embs = np.vstack(embeddings).astype("float32")
        dim = self.resume_embs.shape[1]
        self.index = faiss.IndexFlatIP(dim)
        self.index.add(self.resume_embs)
        logger.info(f"FAISS: {self.resume_embs.shape[0]} vectors")

    async def retrieve_top_k(self, query_text: str, k: int) -> List[Dict]:
        query_vec = await self._openai_embedding(query_text)
        norm = np.linalg.norm(query_vec)
        if norm > 0:
            query_vec /= norm
        sims, indices = self.index.search(np.expand_dims(query_vec, 0), k)
        return [{"idx": int(idx), "similarity": float(sim), "text": self.resume_chunks[idx]}
                for sim, idx in zip(sims[0], indices[0]) if idx >= 0]

    def create_criterion_query(self, criterion_key: str) -> str:
        criteria = self.jd_criteria or {}
        queries = {
            "title_match": f"Job title: {criteria.get('title', 'Unknown')}. Relevant roles.",
            "level_seniority": f"Seniority: {criteria.get('level', 'Not specified')}. Leadership.",
            "experience_years": f"Required: {criteria.get('experience_years', 0)} years experience.",
            "technical_skills": f"Skills: {', '.join(list(self.jd_skills)[:20])}",
            "soft_skills": "Communication, leadership, teamwork, problem-solving.",
            "qualifications": f"Education: {'; '.join(criteria.get('qualifications', [])[:3])}",
            "responsibilities_alignment": f"Responsibilities: {' '.join(criteria.get('responsibilities', [])[:5])}",
            "domain_expertise": f"Domain: {', '.join(criteria.get('domain', ['general']))}"
        }
        return queries.get(criterion_key, self.jd_text[:500])

    def _get_enhanced_cot_prompt(self, criterion: str, query: str, passages: str) -> str:
        skill_overlap = len(self.jd_skills & self.resume_skills)
        skill_overlap_pct = (skill_overlap / len(self.jd_skills) * 100) if self.jd_skills else 0

        return f"""EXPERT RECRUITER EVALUATION

CRITERION: {criterion}
QUERY: {query}

RESUME EVIDENCE:
{passages}


EVALUATE:
1. Extract evidence (direct/transferable/gaps). Be strict: only direct matches get high scores; transferable skills score medium; mismatches or gaps score low.
2. Assess relevance and recency
3. Synthesize holistically

OUTPUT:
PASSAGE_ANALYSIS: [for each passage]
- Evidence: [specific facts]
- Reasoning: [why relevant]
PASSAGE_RELEVANCE_SCORE: [0-100]


CONTEXT:
- JD Skills: {len(self.jd_skills)} | Resume Skills: {len(self.resume_skills)}
- Overlap: {skill_overlap} ({skill_overlap_pct:.1f}%)
- Exp Required: {self.jd_criteria.get('experience_years', 'N/A')} years

**SCORE THIS CRITERION (0-100):**
90-100: Exceeds ✓ (perfect direct match, exceeds requirements)
75-89: Strong ✓ (strong direct match, minor gaps)
60-74: Adequate ✓ (acceptable with some transferable skills)
40-59: Partial ⚠ (partial match, significant gaps)
20-39: Weak ⚠ (weak relevance, major mismatches)
0-19: No fit ✗ (no relevance or complete mismatch)

OUTPUT EXACTLY:
FINAL_CRITERION_SCORE: [0-100]

RECOMMENDATIONS:
- [3-5 bullet points on improvements or strengths, tailored to this criterion. If high score, highlight strengths; if low, suggest specific ways to improve resume or skills for better fit.]

JD REFERENCE:
{self.jd_text[:1500]}"""

# ---------- LLM Helpers ----------
async def _openai_chat_with_retry(session: aiohttp.ClientSession, api_key: str, messages: list,
                                  model: str, timeout: int = 120, retries: int = 3,
                                  temperature: float = 0.1, max_tokens: int = 1024) -> dict:
    endpoint = "https://api.openai.com/v1/chat/completions"
    headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}
    payload = {"model": model, "messages": messages, "temperature": temperature, "max_tokens": max_tokens}

    for attempt in range(retries):
        try:
            async with session.post(endpoint, headers=headers, json=payload, timeout=timeout) as r:
                if r.status != 200:
                    raise RuntimeError(f"API {r.status}")
                return await r.json()
        except Exception as e:
            if attempt == retries - 1:
                raise
            await asyncio.sleep(2 ** attempt)

def parse_final_score_from_text(text: str) -> Optional[int]:
    m = re.search(r"FINAL_CRITERION_SCORE\s*:\s*(\d{1,3})", text)
    if m:
        return max(0, min(100, int(m.group(1))))
    m2 = re.search(r"\b([0-9]{1,3})\b", text)
    if m2:
        s = int(m2.group(1))
        if 0 <= s <= 100:
            return s
    return None

# ---------- 🎯 MAIN FLOW ----------
async def main_flow(resume_path_or_text: Optional[str] = None,
                    jd_url_or_text: Optional[str] = None,
                    config: Optional[Config] = None,
                    openai_api_key: Optional[str] = None,
                    show_verbose: bool = True) -> Dict[str, Any]:

    if config is None:
        config = Config()
    openai_api_key = get_openai_api_key_interactive(openai_api_key)
    matcher = ResumeJDMatcher(config, openai_api_key)

    try:
        session = await matcher.ensure_session()

        # 🚀 INTERACTIVE INPUTS
        if resume_path_or_text is None:
            resume_path_or_text = prompt_for_resume_input()
        if jd_url_or_text is None:
            jd_url_or_text = prompt_for_jd_input()

        await matcher.load_resume(resume_path_or_text)
        await matcher.load_jd(jd_url_or_text)
        matcher.extract_jd_criteria()
        await matcher.embed_resume()

        criterion_results = {}
        for crit in matcher.config.criteria_keys:
            query = matcher.create_criterion_query(crit)
            passages = await matcher.retrieve_top_k(query, matcher.config.top_k_retrieve)

            passages_text = "\n\n".join([f"Passage {i+1}: {p['text']}"
                                       for i, p in enumerate(passages)])
            cot_prompt = matcher._get_enhanced_cot_prompt(crit, query, passages_text)

            messages = [{"role": "system", "content": "Expert recruiter."},
                       {"role": "user", "content": cot_prompt}]

            chat_resp = await _openai_chat_with_retry(session, openai_api_key, messages,
                                                    matcher.config.llm_model,
                                                    max_tokens=matcher.config.max_new_tokens)
            llm_text = chat_resp["choices"][0]["message"]["content"]
            llm_score = parse_final_score_from_text(llm_text)

            avg_sim = np.mean([p["similarity"] for p in passages]) * 100 if passages else 0
            emb_score = float(avg_sim)
            combined_score = (config.weight_llm * (llm_score if llm_score is not None else emb_score) + config.weight_emb * emb_score)
            combined_score = max(0, min(100, combined_score))

            criterion_results[crit] = {
                "llm_score": llm_score, "emb_score": round(emb_score, 2),
                "combined_score": round(combined_score, 2),
                "explanation": llm_text
            }

            if show_verbose:
                logger.info(f"{crit}: {combined_score:.1f} (LLM:{llm_score}, Emb:{emb_score:.1f})")

        overall_score = np.mean([v["combined_score"] for v in criterion_results.values()])

        result = {
            "overall_score": round(overall_score, 2),
            "criteria": criterion_results,
            "jd_criteria": matcher.jd_criteria,
            "resume_skills": sorted(list(matcher.resume_skills))[:50],
            "jd_skills": sorted(list(matcher.jd_skills))[:50]
        }

        print(f"\n🎉 FINAL SCORE: {result['overall_score']:.1f}/100")
        for k, v in result["criteria"].items():
            print(f"  {k}: LLM {v['llm_score']}, EMB {v['emb_score']:.1f}, Average {v['combined_score']:.1f}")
            print(f"    Explanation and Recommendations:\n{v['explanation']}\n")

        print("\nSummary Table:")
        print("| Criterion | LLM | EMB | Average |")
        print("|-----------|-----|-----|---------|")
        for k, v in result["criteria"].items():
            print(f"| {k} | {v['llm_score']} | {v['emb_score']:.1f} | {v['combined_score']:.1f} |")

        return result

    finally:
        await matcher.close()

# ---------- RUN (portable: script + notebook friendly) ----------
if __name__ == "__main__":
    try:
        # Try the simple, normal case first
        try:
            asyncio.run(main_flow())
        except RuntimeError as re:
            # Happens when running inside an existing event loop (e.g. Jupyter/Colab)
            logger.warning("Detected running event loop; attempting notebook-friendly execution.")
            try:
                # Prefer using nest_asyncio if available
                import nest_asyncio
                nest_asyncio.apply()
                loop = asyncio.get_event_loop()
                loop.run_until_complete(main_flow())
            except Exception:
                # If nest_asyncio is not available or something else failed,
                # provide clear instruction for notebook execution.
                logger.error("Could not auto-run inside the existing event loop.")
                logger.info("If you're in a Jupyter/Colab notebook, run: `await main_flow()`")
                raise
    except Exception as e:
        logger.exception(f"Failed: {e}")

# ---------- SAFE COROUTINE CLEANUP ----------
# Add this at the very end of main_flow() definition
# to avoid ignored coroutine warnings
async def main_flow():
    try:
        # ... your existing main_flow code ...
        pass  # replace with actual code
    finally:
        # Ensure all pending tasks are awaited
        pending = [t for t in asyncio.all_tasks() if not t.done()]
        if pending:
            await asyncio.gather(*pending, return_exceptions=True)
```

```python
"""
Advanced Hybrid RAG + LLM Reranker for Resume ↔ JD Matching (ENHANCED UPLOAD VERSION)

🎯 FEATURES:
- 📁 UPLOAD .docx Resume (Colab button / Desktop picker / Paste fallback)
- 🔗 JD from URL or Paste
- 🧠 GPT-4o Reranking + Embeddings
- 📊 Detailed Score Breakdown
- ✅ Interactive & Production Ready

Usage: python resume_jd_matcher_fixed.py
"""

import os
import re
import json
import math
import time
import asyncio
import logging
from typing import List, Dict, Tuple, Optional, Set, Any
from collections import Counter
import numpy as np
import pandas as pd
from docx import Document
from rapidfuzz import process, fuzz
import spacy
from tqdm.auto import tqdm
import aiohttp
from bs4 import BeautifulSoup
import faiss
import html
from sklearn.feature_extraction.text import TfidfVectorizer
import getpass

# ---------- Logging ----------
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
logger = logging.getLogger(__name__)

# ---------- Load spaCy ----------
try:
    nlp = spacy.load("en_core_web_lg")
except Exception as e:
    logger.warning("spaCy model not found. Run: python -m spacy download en_core_web_lg")
    raise

# ---------- Configuration ----------
class Config:
    def __init__(self):
        self.embed_model = "text-embedding-3-large"
        self.llm_model = "gpt-4o"
        self.chunk_size = 150
        self.chunk_overlap = 30
        self.top_k_retrieve = 6
        self.weight_llm = 0.75
        self.weight_emb = 0.25
        self.max_new_tokens = 1200
        self.temperature = 0.1
        self.api_timeout = 120
        self.api_retries = 3
        self.concurrent_embeddings = 6
        self.criteria_keys = ["title_match", "level_seniority", "experience_years",
                             "technical_skills", "soft_skills", "qualifications",
                             "responsibilities_alignment", "domain_expertise"]
        self.redact_pii = False
        self.min_skill_length = 2
        self.max_skill_length = 50
        self.tfidf_top_n = 100

# ---------- 🚀 FIXED AUTOMATED UPLOAD FUNCTIONS ----------
import tempfile
import sys
import time
from pathlib import Path

def safe_filename(filename: str) -> str:
    """Clean filename for temp storage"""
    return re.sub(r'[<>:"/\\|?*]', '_', filename)

def prompt_for_resume_input() -> str:
    """🔥 FULLY AUTOMATED: Detects & handles Colab/Jupyter/Desktop in ONE GO"""
    print("\n📁 RESUME UPLOAD (AUTO-DETECTED)")

    # 1) COLAB - IMMEDIATE UPLOAD
    if "google.colab" in sys.modules:
        try:
            from google.colab import files
            print("✅ COLAB DETECTED - UPLOADING...")
            uploaded = files.upload()
            if uploaded:
                fname = next(iter(uploaded.keys()))
                dest = os.path.join(tempfile.gettempdir(), safe_filename(fname))
                with open(dest, "wb") as f:
                    f.write(uploaded[fname])
                print(f"✅ SAVED: {dest}")
                return dest
        except:
            pass

    # 2) JUPYTER - IMMEDIATE WIDGET
    if "ipykernel" in sys.modules:
        try:
            import ipywidgets as widgets
            from IPython.display import display
            print("✅ JUPYTER DETECTED - UPLOAD BELOW:")
            uploader = widgets.FileUpload(accept='.docx', multiple=False)
            display(uploader)
            # AUTO-WAIT 10s MAX
            for _ in range(20):  # 10 seconds
                if uploader.value:
                    uploaded_value = uploader.value
                    if isinstance(uploaded_value, dict):
                        name = list(uploaded_value.keys())[0]
                        content = uploaded_value[name]['content']
                    else:
                        item = uploaded_value[0]
                        name = item.get('name') or f"resume_{int(time.time())}.docx"
                        content = item.get('content')
                    dest = os.path.join(tempfile.gettempdir(), safe_filename(name))
                    with open(dest, "wb") as f:
                        f.write(content)
                    print(f"✅ JUPYTER SAVED: {dest}")
                    return dest
                time.sleep(0.5)
            print("⚠️ JUPYTER TIMEOUT - FALLBACK")
        except:
            pass

    # 3) DESKTOP - ONE-CLICK DIALOG
    try:
        import tkinter as tk
        from tkinter import filedialog
        root = tk.Tk()
        root.withdraw()
        print("✅ DESKTOP DETECTED - SELECT FILE")
        path = filedialog.askopenfilename(
            title="Select Resume (.docx)",
            filetypes=[("Word", "*.docx"), ("All", "*.*")]
        )
        if path and os.path.exists(path):
            print(f"✅ DESKTOP LOADED: {path}")
            return path
        print("⚠️ NO FILE - FALLBACK")
    except:
        pass

    # 4) ULTRA-FAST CLI (2 seconds total)
    print("⚡ QUICK: Enter path or press Enter:")
    path = input().strip()
    if path and os.path.exists(path):
        return path

    print("📝 PASTE RESUME (2 lines max):")
    text = input().strip()
    if text:
        return text

    raise RuntimeError("❌ No resume provided")

def prompt_for_jd_input() -> str:
    """JD input: URL or paste"""
    print("\n" + "="*50)
    print("📋 JOB DESCRIPTION INPUT:")
    print("="*50)
    print(" - URL: Paste LinkedIn/job URL (starts with http)")
    print(" - TEXT: Press Enter then paste JD")

    first = input("Enter JD URL or press Enter to paste: ").strip()
    if first:
        return first

    print("Paste JD text (end with blank line):")
    lines = []
    while True:
        try:
            line = input()
            if not line.strip():
                break
            lines.append(line)
        except EOFError:
            break
    text = "\n".join(lines).strip()
    if not text:
        raise RuntimeError("No JD provided")
    return text

def get_openai_api_key_interactive(provided_key: Optional[str] = None) -> str:
    """Get API key: provided → env → prompt"""
    if provided_key:
        return provided_key
    env_key = os.environ.get("OPENAI_API_KEY") or os.environ.get("OPENAI_APIKEY")
    if env_key:
        logger.info("Using OPENAI_API_KEY from environment")
        return env_key

    try:
        key = getpass.getpass("Enter OpenAI API key: ")
    except:
        key = input("Enter OpenAI API key: ").strip()

    if not key:
        raise RuntimeError("API key required")
    return key

# ---------- PII Redaction ----------
def redact_pii(text: str, nlp_model) -> str:
    if not text:
        return text
    doc = nlp_model(text)
    out = []
    last = 0
    for ent in doc.ents:
        if ent.label_ in {"PERSON", "GPE", "ORG"} or re.match(r".+@.+\..+", ent.text):
            out.append(text[last:ent.start_char])
            out.append(f"[REDACTED_{ent.label_}]")
            last = ent.end_char
    out.append(text[last:])
    return re.sub(r"\s{2,}", " ", "".join(out)).strip()

# ---------- Resume Parser ----------
class ResumeParser:
    SECTION_HEADERS = {
        "experience": r"(work\s+)?experience|employment\s+history|professional\s+background",
        "education": r"education|academic\s+background|qualifications",
        "skills": r"skills|technical\s+skills|competencies|expertise",
        "summary": r"summary|profile|objective|about\s+me",
        "projects": r"projects|portfolio",
        "certifications": r"certifications?|licenses?|credentials",
        "achievements": r"achievements?|awards?|honors?"
    }

    def __init__(self, text: str, nlp_model):
        self.text = text
        self.nlp = nlp_model
        self.sections = {}
        self._parse_sections()

    def _parse_sections(self):
        lines = self.text.split("\n")
        current_section = "header"
        current_content = []

        for line in lines:
            line_stripped = line.strip()
            if not line_stripped:
                continue

            matched_section = None
            for section_name, pattern in self.SECTION_HEADERS.items():
                if re.match(pattern, line_stripped, re.I) and len(line_stripped.split()) <= 5:
                    matched_section = section_name
                    break

            if matched_section:
                if current_content:
                    self.sections[current_section] = "\n".join(current_content)
                current_section = matched_section
                current_content = []
            else:
                current_content.append(line_stripped)

        if current_content:
            self.sections[current_section] = "\n".join(current_content)

        logger.info(f"Detected sections: {list(self.sections.keys())}")

    def get_section(self, section_name: str) -> str:
        return self.sections.get(section_name, "")

    def get_full_text(self) -> str:
        return self.text

# ---------- Dynamic Skills Extractor ----------
class DynamicSkillsExtractor:
    def __init__(self, nlp_model):
        self.nlp = nlp_model
        self.common_stop_words = {
            "experience", "work", "team", "project", "company", "role", "position",
            "responsible", "including", "using", "related", "various", "multiple",
            "required", "preferred", "ability", "knowledge", "understanding"
        }

    def extract_from_text(self, text: str, top_n: int = 100) -> Set[str]:
        skills = set()
        skills.update(self._extract_via_ner(text))
        skills.update(self._extract_via_tfidf(text, top_n))
        skills.update(self._extract_via_patterns(text))
        filtered = self._filter_skills(skills)
        logger.info(f"Extracted {len(filtered)} skills")
        return filtered

    def _extract_via_ner(self, text: str) -> Set[str]:
        doc = self.nlp(text)
        skills = set()
        for ent in doc.ents:
            if ent.label_ in {"PRODUCT", "ORG", "NORP", "LANGUAGE", "WORK_OF_ART"}:
                skills.add(ent.text.strip())
        for chunk in doc.noun_chunks:
            text_lower = chunk.text.lower()
            if any(indicator in text_lower for indicator in
                   ["software", "tool", "framework", "language", "platform", "system"]):
                skills.add(chunk.text.strip())
        return skills

    def _extract_via_tfidf(self, text: str, top_n: int) -> Set[str]:
        try:
            vectorizer = TfidfVectorizer(
                max_features=top_n, ngram_range=(1, 3), stop_words='english',
                token_pattern=r'\b[A-Za-z][A-Za-z0-9\.\+\#\-]*\b'
            )
            tfidf_matrix = vectorizer.fit_transform([text])
            feature_names = vectorizer.get_feature_names_out()
            scores = tfidf_matrix.toarray()[0]
            top_indices = scores.argsort()[-top_n:][::-1]
            return {feature_names[i] for i in top_indices if scores[i] > 0.01}
        except:
            return set()

    def _extract_via_patterns(self, text: str) -> Set[str]:
        skills = set()
        skills.update(re.findall(r'\b([A-Z][a-z]+(?:[A-Z][a-z]+)*(?:\.[a-z]+)?)\b', text))
        skills.update(re.findall(r'\b([A-Z]{2,6})\b', text))
        potential = re.findall(r'\b([A-Za-z]+(?:\s+\d+(?:\.\d+)?)?)\b', text)
        skills.update([p for p in potential if any(c.isdigit() for c in p)])
        skills.update(re.findall(r'\b([A-Za-z]+-[A-Za-z]+(?:-[A-Za-z]+)?)\b', text))
        return skills

    def _filter_skills(self, skills: Set[str]) -> Set[str]:
        filtered = set()
        for skill in skills:
            skill_clean = skill.strip()
            skill_lower = skill_clean.lower()
            if (2 <= len(skill_clean) <= 50 and
                skill_lower not in self.common_stop_words and
                re.search(r'[A-Za-z]', skill_clean) and
                skill_lower not in {"the", "and", "for", "with", "from", "this", "that", "will", "can"}):
                filtered.add(skill_clean)
        return filtered

# ---------- JD Extraction ----------
async def load_jd_from_url(url: str, session: aiohttp.ClientSession, timeout: int = 30) -> str:
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    }
    async with session.get(url, headers=headers, timeout=timeout, allow_redirects=True) as r:
        r.raise_for_status()
        html_text = await r.text()

    soup = BeautifulSoup(html_text, "html.parser")
    linkedin_selectors = [
        {"name": "div", "attrs": {"class": "description__text"}},
        {"name": "div", "attrs": {"class": "show-more-less-html__markup"}},
        {"name": "section", "attrs": {"class": "description"}},
    ]
    generic_selectors = [
        {"name": "div", "attrs": {"class": re.compile(r"(?i)(job[-_]?description|description)")}},
        {"name": "section", "attrs": {"class": re.compile(r"(?i)(job|description)")}},
    ]

    all_selectors = linkedin_selectors + generic_selectors
    text = ""
    for sel in all_selectors:
        node = soup.find(sel.get("name"), sel.get("attrs"))
        if node:
            text = node.get_text(separator="\n", strip=True)
            if len(text.split()) >= 50:
                break

    if not text or len(text.split()) < 50:
        paragraphs = soup.find_all("p")
        text = "\n".join([p.get_text(strip=True) for p in paragraphs])

    return re.sub(r"\n{3,}", "\n\n", html.unescape(text)).strip()

def read_docx(path: str) -> str:
    doc = Document(path)
    paragraphs = []
    for p in doc.paragraphs:
        text = p.text.strip()
        if text:
            paragraphs.append(text)
    for table in doc.tables:
        for row in table.rows:
            for cell in row.cells:
                text = cell.text.strip()
                if text:
                    paragraphs.append(text)
    return "\n".join(paragraphs)

def extract_experience_years(text: str) -> Optional[int]:
    patterns = [
        r"(\d+)\+?\s*(?:years?|yrs?)[\s\w]*(?:of\s+)?(?:experience|exp)",
        r"(?:minimum|at least)\s+(\d+)\s*(?:years?|yrs?)",
        r"(\d+)(?:\s*-\s*(\d+))?\s*(?:years?|yrs?)",
    ]
    max_years = None
    for pattern in patterns:
        matches = re.findall(pattern, text, re.I)
        for match in matches:
            nums = [int(n) for n in match if n and n.isdigit()]
            if nums:
                year = max(nums)
                if max_years is None or year > max_years:
                    max_years = year
    return max_years

# ---------- Main Matcher ----------
class ResumeJDMatcher:
    def __init__(self, config: Config, openai_api_key: str):
        self.config = config
        self.openai_api_key = openai_api_key  # ✅ BUGFIX: actually stores key
        self.resume_text: Optional[str] = None
        self.resume_parser: Optional[ResumeParser] = None
        self.jd_text: Optional[str] = None
        self.jd_criteria: Optional[Dict] = None
        self.jd_skills: Set[str] = set()
        self.resume_skills: Set[str] = set()
        self.resume_chunks: List[str] = []
        self.index: Optional[faiss.Index] = None
        self.resume_embs: Optional[np.ndarray] = None
        self._aiohttp_session: Optional[aiohttp.ClientSession] = None
        self._embed_semaphore = asyncio.Semaphore(self.config.concurrent_embeddings)
        self.skills_extractor = DynamicSkillsExtractor(nlp)

    async def ensure_session(self) -> aiohttp.ClientSession:
        if self._aiohttp_session is None:
            self._aiohttp_session = aiohttp.ClientSession()
        return self._aiohttp_session

    async def close(self):
        if self._aiohttp_session:
            await self._aiohttp_session.close()

    async def load_resume(self, path_or_text: str) -> None:
        if path_or_text.endswith(".docx") and os.path.exists(path_or_text):
            self.resume_text = read_docx(path_or_text)
            logger.info(f"✅ Loaded .docx: {path_or_text}")
        else:
            self.resume_text = path_or_text

        if not self.resume_text.strip():
            raise ValueError("Empty resume")

        self.resume_parser = ResumeParser(self.resume_text, nlp)
        self.resume_skills = self.skills_extractor.extract_from_text(self.resume_text)
        logger.info(f"Resume: {len(self.resume_text.split())} words, {len(self.resume_skills)} skills")

    async def load_jd(self, url_or_text: str) -> None:
        session = await self.ensure_session()
        if url_or_text.startswith("http"):
            self.jd_text = await load_jd_from_url(url_or_text, session)
            logger.info(f"✅ JD from URL: {len(self.jd_text.split())} words")
        else:
            self.jd_text = url_or_text

        if not self.jd_text.strip():
            raise ValueError("Empty JD")

        self.jd_skills = self.skills_extractor.extract_from_text(self.jd_text)
        logger.info(f"JD: {len(self.jd_text.split())} words, {len(self.jd_skills)} skills")

    def extract_jd_criteria(self) -> None:
        doc = nlp(self.jd_text)
        title = ""
        for line in self.jd_text.split("\n")[:15]:
            line_clean = line.strip()
            if 3 <= len(line_clean.split()) <= 12 and not line_clean.endswith(":"):
                title = line_clean
                break

        level_keywords = ["senior", "junior", "lead", "principal", "staff", "manager"]
        level = next((token.text for sent in list(doc.sents)[:20]
                     for token in sent if token.lower_ in level_keywords), "")

        exp_years = extract_experience_years(self.jd_text)
        qual_keywords = ["bachelor", "master", "phd", "degree", "certification"]
        quals = [sent.text.strip() for sent in doc.sents
                if any(kw in sent.text.lower() for kw in qual_keywords)][:10]

        action_verbs = {"develop", "design", "lead", "manage", "implement", "build"}
        resps = [sent.text.strip() for sent in doc.sents
                if any(token.lemma_ in action_verbs for token in sent)][:15]

        domain_keywords = ["healthcare", "medical", "dentistry", "dental", "finance", "fintech", "ecommerce", "saas", "cloud", "ai", "machine learning", "data science", "education", "retail", "human resources", "hr", "marketing", "sales", "engineering", "software development", "manufacturing", "law", "legal", "hospitality", "energy", "environment", "government", "media", "entertainment", "transportation", "logistics"]
        domain = [kw for kw in domain_keywords if kw in self.jd_text.lower()]

        self.jd_criteria = {
            "title": title, "level": level, "experience_years": exp_years,
            "skills": sorted(list(self.jd_skills))[:50],
            "qualifications": quals, "responsibilities": resps, "domain": domain
        }

    def chunk_resume(self) -> None:
        doc = nlp(self.resume_text)
        chunks = []
        current_chunk = []
        current_words = 0

        for sent in doc.sents:
            sent_text = sent.text.strip()
            sent_words = len(sent_text.split())
            if current_words + sent_words > self.config.chunk_size and current_chunk:
                chunk_text = " ".join(current_chunk)
                chunks.append(chunk_text)
                current_chunk = []
                current_words = 0
            current_chunk.append(sent_text)
            current_words += sent_words

        if current_chunk:
            chunks.append(" ".join(current_chunk))

        self.resume_chunks = chunks
        logger.info(f"Chunked into {len(chunks)} segments")

    async def _openai_embedding(self, text: str) -> np.ndarray:
        session = await self.ensure_session()
        endpoint = "https://api.openai.com/v1/embeddings"
        headers = {"Authorization": f"Bearer {self.openai_api_key}", "Content-Type": "application/json"}
        payload = {"input": text[:8000], "model": self.config.embed_model}

        async with session.post(endpoint, headers=headers, json=payload,
                               timeout=self.config.api_timeout) as r:
            data = await r.json()
        vec = np.array(data["data"][0]["embedding"], dtype=np.float32)
        norm = np.linalg.norm(vec)
        if norm > 0:
            vec /= norm
        return vec

    async def embed_resume(self) -> None:
        if not self.resume_chunks:
            self.chunk_resume()

        tasks = [self._openai_embedding(chunk) for chunk in self.resume_chunks]
        embeddings = []
        for task in tqdm(asyncio.as_completed(tasks), total=len(tasks), desc="Embedding"):
            embeddings.append(await task)

        self.resume_embs = np.vstack(embeddings).astype("float32")
        dim = self.resume_embs.shape[1]
        self.index = faiss.IndexFlatIP(dim)
        self.index.add(self.resume_embs)
        logger.info(f"FAISS: {self.resume_embs.shape[0]} vectors")

    async def retrieve_top_k(self, query_text: str, k: int) -> List[Dict]:
        query_vec = await self._openai_embedding(query_text)
        norm = np.linalg.norm(query_vec)
        if norm > 0:
            query_vec /= norm
        sims, indices = self.index.search(np.expand_dims(query_vec, 0), k)
        return [{"idx": int(idx), "similarity": float(sim), "text": self.resume_chunks[idx]}
                for sim, idx in zip(sims[0], indices[0]) if idx >= 0]

    def create_criterion_query(self, criterion_key: str) -> str:
        criteria = self.jd_criteria or {}
        queries = {
            "title_match": f"Job title: {criteria.get('title', 'Unknown')}. Relevant roles.",
            "level_seniority": f"Seniority: {criteria.get('level', 'Not specified')}. Leadership.",
            "experience_years": f"Required: {criteria.get('experience_years', 0)} years experience.",
            "technical_skills": f"Skills: {', '.join(list(self.jd_skills)[:20])}",
            "soft_skills": "Communication, leadership, teamwork, problem-solving.",
            "qualifications": f"Education: {'; '.join(criteria.get('qualifications', [])[:3])}",
            "responsibilities_alignment": f"Responsibilities: {' '.join(criteria.get('responsibilities', [])[:5])}",
            "domain_expertise": f"Domain: {', '.join(criteria.get('domain', ['general']))}"
        }
        return queries.get(criterion_key, self.jd_text[:500])

    def _get_enhanced_cot_prompt(self, criterion: str, query: str, passages: str) -> str:
        skill_overlap = len(self.jd_skills & self.resume_skills)
        skill_overlap_pct = (skill_overlap / len(self.jd_skills) * 100) if self.jd_skills else 0

        return f"""EXPERT RECRUITER EVALUATION - STRICT JD ALIGNMENT

You must evaluate how well the resume evidence aligns with the Job Description (JD) requirements for this criterion. Score ONLY based on direct matches to the JD's specific needs. Ignore any resume strengths that do not relate to the JD. Be extremely strict: if the resume does not demonstrate expertise in the JD's domain or field, score very low, even if the resume shows strong skills in unrelated areas.

CRITERION: {criterion}
QUERY: {query}

RESUME EVIDENCE:
{passages}


EVALUATE:
1. For each passage, extract evidence relevant ONLY to the JD (direct/transferable/gaps). Be strict: only evidence directly matching JD gets high passage scores; transferable to JD medium; unrelated or gaps low.
2. Assess relevance to JD, recency, and depth.
3. Synthesize holistically for overall JD match.

OUTPUT:
PASSAGE_ANALYSIS: [for each passage]
- Evidence: [specific facts relevant to JD]
- Reasoning: [why it matches or doesn't match the JD]
PASSAGE_RELEVANCE_SCORE: [0-100]


CONTEXT:
- JD Skills: {len(self.jd_skills)} | Resume Skills: {len(self.resume_skills)}
- Overlap: {skill_overlap} ({skill_overlap_pct:.1f}%)
- Exp Required: {self.jd_criteria.get('experience_years', 'N/A')} years

**SCORE THIS CRITERION (0-100) BASED ON JD ALIGNMENT:**
90-100: Exceeds JD requirements ✓ (perfect direct match, exceeds JD requirements)
75-89: Strong match to JD ✓ (strong direct match, minor gaps)
60-74: Adequate for JD ✓ (acceptable with some transferable skills to JD)
40-59: Partial match to JD ⚠ (partial match, significant gaps)
20-39: Weak match to JD ⚠ (weak relevance to JD, major mismatches)
0-19: No fit for JD ✗ (no relevance to JD or complete mismatch)

The FINAL_CRITERION_SCORE is a holistic assessment of JD match, not necessarily an average of passage scores. It can be lower if overall alignment is poor despite some weakly relevant passages.

OUTPUT EXACTLY:
FINAL_CRITERION_SCORE: [0-100]

RECOMMENDATIONS:
- [3-5 bullet points on improvements or strengths, tailored to this criterion and JD. If high score, highlight strengths aligning with JD; if low, suggest specific ways to improve resume or skills for better JD fit.]

JD REFERENCE:
{self.jd_text[:1500]}"""

# ---------- LLM Helpers ----------
async def _openai_chat_with_retry(session: aiohttp.ClientSession, api_key: str, messages: list,
                                  model: str, timeout: int = 120, retries: int = 3,
                                  temperature: float = 0.1, max_tokens: int = 1024) -> dict:
    endpoint = "https://api.openai.com/v1/chat/completions"
    headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}
    payload = {"model": model, "messages": messages, "temperature": temperature, "max_tokens": max_tokens}

    for attempt in range(retries):
        try:
            async with session.post(endpoint, headers=headers, json=payload, timeout=timeout) as r:
                if r.status != 200:
                    raise RuntimeError(f"API {r.status}")
                return await r.json()
        except Exception as e:
            if attempt == retries - 1:
                raise
            await asyncio.sleep(2 ** attempt)

def parse_final_score_from_text(text: str) -> Optional[int]:
    m = re.search(r"FINAL_CRITERION_SCORE\s*:\s*(\d{1,3})", text)
    if m:
        return max(0, min(100, int(m.group(1))))
    m2 = re.search(r"\b([0-9]{1,3})\b", text)
    if m2:
        s = int(m2.group(1))
        if 0 <= s <= 100:
            return s
    return None

# ---------- 🎯 MAIN FLOW ----------
async def main_flow(resume_path_or_text: Optional[str] = None,
                    jd_url_or_text: Optional[str] = None,
                    config: Optional[Config] = None,
                    openai_api_key: Optional[str] = None,
                    show_verbose: bool = True) -> Dict[str, Any]:

    if config is None:
        config = Config()
    openai_api_key = get_openai_api_key_interactive(openai_api_key)
    matcher = ResumeJDMatcher(config, openai_api_key)

    try:
        session = await matcher.ensure_session()

        # 🚀 INTERACTIVE INPUTS
        if resume_path_or_text is None:
            resume_path_or_text = prompt_for_resume_input()
        if jd_url_or_text is None:
            jd_url_or_text = prompt_for_jd_input()

        await matcher.load_resume(resume_path_or_text)
        await matcher.load_jd(jd_url_or_text)
        matcher.extract_jd_criteria()
        await matcher.embed_resume()

        criterion_results = {}
        for crit in matcher.config.criteria_keys:
            query = matcher.create_criterion_query(crit)
            passages = await matcher.retrieve_top_k(query, matcher.config.top_k_retrieve)

            passages_text = "\n\n".join([f"Passage {i+1}: {p['text']}"
                                       for i, p in enumerate(passages)])
            cot_prompt = matcher._get_enhanced_cot_prompt(crit, query, passages_text)

            messages = [{"role": "system", "content": "Expert recruiter."},
                       {"role": "user", "content": cot_prompt}]

            chat_resp = await _openai_chat_with_retry(session, openai_api_key, messages,
                                                    matcher.config.llm_model,
                                                    max_tokens=matcher.config.max_new_tokens)
            llm_text = chat_resp["choices"][0]["message"]["content"]
            llm_score = parse_final_score_from_text(llm_text)

            avg_sim = np.mean([p["similarity"] for p in passages]) * 100 if passages else 0
            emb_score = float(avg_sim)
            combined_score = (config.weight_llm * (llm_score if llm_score is not None else emb_score) + config.weight_emb * emb_score)
            combined_score = max(0, min(100, combined_score))

            criterion_results[crit] = {
                "llm_score": llm_score, "emb_score": round(emb_score, 2),
                "combined_score": round(combined_score, 2),
                "explanation": llm_text
            }

            if show_verbose:
                logger.info(f"{crit}: {combined_score:.1f} (LLM:{llm_score}, Emb:{emb_score:.1f})")

        overall_score = np.mean([v["combined_score"] for v in criterion_results.values()])

        result = {
            "overall_score": round(overall_score, 2),
            "criteria": criterion_results,
            "jd_criteria": matcher.jd_criteria,
            "resume_skills": sorted(list(matcher.resume_skills))[:50],
            "jd_skills": sorted(list(matcher.jd_skills))[:50]
        }

        print(f"\n🎉 FINAL SCORE: {result['overall_score']:.1f}/100")
        for k, v in result["criteria"].items():
            print(f"  {k}: LLM {v['llm_score']}, EMB {v['emb_score']:.1f}, Average {v['combined_score']:.1f}")
            print(f"    Explanation and Recommendations:\n{v['explanation']}\n")

        print("\nSummary Table:")
        print("| Criterion | LLM | EMB | Average |")
        print("|-----------|-----|-----|---------|")
        for k, v in result["criteria"].items():
            print(f"| {k} | {v['llm_score']} | {v['emb_score']:.1f} | {v['combined_score']:.1f} |")

        return result

    finally:
        await matcher.close()

# ---------- RUN (portable: script + notebook friendly) ----------
if __name__ == "__main__":
    try:
        # Try the simple, normal case first
        try:
            asyncio.run(main_flow())
        except RuntimeError as re:
            # Happens when running inside an existing event loop (e.g. Jupyter/Colab)
            logger.warning("Detected running event loop; attempting notebook-friendly execution.")
            try:
                # Prefer using nest_asyncio if available
                import nest_asyncio
                nest_asyncio.apply()
                loop = asyncio.get_event_loop()
                loop.run_until_complete(main_flow())
            except Exception:
                # If nest_asyncio is not available or something else failed,
                # provide clear instruction for notebook execution.
                logger.error("Could not auto-run inside the existing event loop.")
                logger.info("If you're in a Jupyter/Colab notebook, run: `await main_flow()`")
                raise
    except Exception as e:
        logger.exception(f"Failed: {e}")
```

```python
"""
Advanced Hybrid RAG + LLM Reranker for Resume ↔ JD Matching (ENHANCED UPLOAD VERSION)

🎯 FEATURES:
- 📁 UPLOAD .docx Resume (Colab button / Desktop picker / Paste fallback)
- 🔗 JD from URL or Paste
- 🧠 GPT-4o Reranking + Embeddings
- 📊 Detailed Score Breakdown
- ✅ Interactive & Production Ready

Usage: python resume_jd_matcher_fixed.py
"""

import os
import re
import json
import math
import time
import asyncio
import logging
from typing import List, Dict, Tuple, Optional, Set, Any
from collections import Counter
import numpy as np
import pandas as pd
from docx import Document
from rapidfuzz import process, fuzz
import spacy
from tqdm.auto import tqdm
import aiohttp
from bs4 import BeautifulSoup
import faiss
import html
from sklearn.feature_extraction.text import TfidfVectorizer
import getpass

# ---------- Logging ----------
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
logger = logging.getLogger(__name__)

# ---------- Load spaCy ----------
try:
    nlp = spacy.load("en_core_web_lg")
except Exception as e:
    logger.warning("spaCy model not found. Run: python -m spacy download en_core_web_lg")
    raise

# ---------- Configuration ----------
class Config:
    def __init__(self):
        self.embed_model = "text-embedding-3-large"
        self.llm_model = "gpt-4o"
        self.chunk_size = 150
        self.chunk_overlap = 30
        self.top_k_retrieve = 6
        self.weight_llm = 0.80
        self.weight_emb = 0.20
        self.max_new_tokens = 1200
        self.temperature = 0.1
        self.api_timeout = 120
        self.api_retries = 3
        self.concurrent_embeddings = 6
        self.criteria_keys = ["title_match", "level_seniority", "experience_years",
                             "technical_skills", "soft_skills", "qualifications",
                             "responsibilities_alignment", "domain_expertise"]
        self.redact_pii = False
        self.min_skill_length = 2
        self.max_skill_length = 50
        self.tfidf_top_n = 100

# ---------- 🚀 FIXED AUTOMATED UPLOAD FUNCTIONS ----------
import tempfile
import sys
import time
from pathlib import Path

def safe_filename(filename: str) -> str:
    """Clean filename for temp storage"""
    return re.sub(r'[<>:"/\\|?*]', '_', filename)

def prompt_for_resume_input() -> str:
    """🔥 FULLY AUTOMATED: Detects & handles Colab/Jupyter/Desktop in ONE GO"""
    print("\n📁 RESUME UPLOAD (AUTO-DETECTED)")

    # 1) COLAB - IMMEDIATE UPLOAD
    if "google.colab" in sys.modules:
        try:
            from google.colab import files
            print("✅ COLAB DETECTED - UPLOADING...")
            uploaded = files.upload()
            if uploaded:
                fname = next(iter(uploaded.keys()))
                dest = os.path.join(tempfile.gettempdir(), safe_filename(fname))
                with open(dest, "wb") as f:
                    f.write(uploaded[fname])
                print(f"✅ SAVED: {dest}")
                return dest
        except:
            pass

    # 2) JUPYTER - IMMEDIATE WIDGET
    if "ipykernel" in sys.modules:
        try:
            import ipywidgets as widgets
            from IPython.display import display
            print("✅ JUPYTER DETECTED - UPLOAD BELOW:")
            uploader = widgets.FileUpload(accept='.docx', multiple=False)
            display(uploader)
            # AUTO-WAIT 10s MAX
            for _ in range(20):  # 10 seconds
                if uploader.value:
                    uploaded_value = uploader.value
                    if isinstance(uploaded_value, dict):
                        name = list(uploaded_value.keys())[0]
                        content = uploaded_value[name]['content']
                    else:
                        item = uploaded_value[0]
                        name = item.get('name') or f"resume_{int(time.time())}.docx"
                        content = item.get('content')
                    dest = os.path.join(tempfile.gettempdir(), safe_filename(name))
                    with open(dest, "wb") as f:
                        f.write(content)
                    print(f"✅ JUPYTER SAVED: {dest}")
                    return dest
                time.sleep(0.5)
            print("⚠️ JUPYTER TIMEOUT - FALLBACK")
        except:
            pass

    # 3) DESKTOP - ONE-CLICK DIALOG
    try:
        import tkinter as tk
        from tkinter import filedialog
        root = tk.Tk()
        root.withdraw()
        print("✅ DESKTOP DETECTED - SELECT FILE")
        path = filedialog.askopenfilename(
            title="Select Resume (.docx)",
            filetypes=[("Word", "*.docx"), ("All", "*.*")]
        )
        if path and os.path.exists(path):
            print(f"✅ DESKTOP LOADED: {path}")
            return path
        print("⚠️ NO FILE - FALLBACK")
    except:
        pass

    # 4) ULTRA-FAST CLI (2 seconds total)
    print("⚡ QUICK: Enter path or press Enter:")
    path = input().strip()
    if path and os.path.exists(path):
        return path

    print("📝 PASTE RESUME (2 lines max):")
    text = input().strip()
    if text:
        return text

    raise RuntimeError("❌ No resume provided")

def prompt_for_jd_input() -> str:
    """JD input: URL or paste"""
    print("\n" + "="*50)
    print("📋 JOB DESCRIPTION INPUT:")
    print("="*50)
    print(" - URL: Paste LinkedIn/job URL (starts with http)")
    print(" - TEXT: Press Enter then paste JD")

    first = input("Enter JD URL or press Enter to paste: ").strip()
    if first:
        return first

    print("Paste JD text (end with blank line):")
    lines = []
    while True:
        try:
            line = input()
            if not line.strip():
                break
            lines.append(line)
        except EOFError:
            break
    text = "\n".join(lines).strip()
    if not text:
        raise RuntimeError("No JD provided")
    return text

def get_openai_api_key_interactive(provided_key: Optional[str] = None) -> str:
    """Get API key: provided → env → prompt"""
    if provided_key:
        return provided_key
    env_key = os.environ.get("OPENAI_API_KEY") or os.environ.get("OPENAI_APIKEY")
    if env_key:
        logger.info("Using OPENAI_API_KEY from environment")
        return env_key

    try:
        key = getpass.getpass("Enter OpenAI API key: ")
    except:
        key = input("Enter OpenAI API key: ").strip()

    if not key:
        raise RuntimeError("API key required")
    return key

# ---------- PII Redaction ----------
def redact_pii(text: str, nlp_model) -> str:
    if not text:
        return text
    doc = nlp_model(text)
    out = []
    last = 0
    for ent in doc.ents:
        if ent.label_ in {"PERSON", "GPE", "ORG"} or re.match(r".+@.+\..+", ent.text):
            out.append(text[last:ent.start_char])
            out.append(f"[REDACTED_{ent.label_}]")
            last = ent.end_char
    out.append(text[last:])
    return re.sub(r"\s{2,}", " ", "".join(out)).strip()

# ---------- Resume Parser ----------
class ResumeParser:
    SECTION_HEADERS = {
        "experience": r"(work\s+)?experience|employment\s+history|professional\s+background",
        "education": r"education|academic\s+background|qualifications",
        "skills": r"skills|technical\s+skills|competencies|expertise",
        "summary": r"summary|profile|objective|about\s+me",
        "projects": r"projects|portfolio",
        "certifications": r"certifications?|licenses?|credentials",
        "achievements": r"achievements?|awards?|honors?"
    }

    def __init__(self, text: str, nlp_model):
        self.text = text
        self.nlp = nlp_model
        self.sections = {}
        self._parse_sections()

    def _parse_sections(self):
        lines = self.text.split("\n")
        current_section = "header"
        current_content = []

        for line in lines:
            line_stripped = line.strip()
            if not line_stripped:
                continue

            matched_section = None
            for section_name, pattern in self.SECTION_HEADERS.items():
                if re.match(pattern, line_stripped, re.I) and len(line_stripped.split()) <= 5:
                    matched_section = section_name
                    break

            if matched_section:
                if current_content:
                    self.sections[current_section] = "\n".join(current_content)
                current_section = matched_section
                current_content = []
            else:
                current_content.append(line_stripped)

        if current_content:
            self.sections[current_section] = "\n".join(current_content)

        logger.info(f"Detected sections: {list(self.sections.keys())}")

    def get_section(self, section_name: str) -> str:
        return self.sections.get(section_name, "")

    def get_full_text(self) -> str:
        return self.text

# ---------- Dynamic Skills Extractor ----------
class DynamicSkillsExtractor:
    def __init__(self, nlp_model):
        self.nlp = nlp_model
        self.common_stop_words = {
            "experience", "work", "team", "project", "company", "role", "position",
            "responsible", "including", "using", "related", "various", "multiple",
            "required", "preferred", "ability", "knowledge", "understanding"
        }

    def extract_from_text(self, text: str, top_n: int = 100) -> Set[str]:
        skills = set()
        skills.update(self._extract_via_ner(text))
        skills.update(self._extract_via_tfidf(text, top_n))
        skills.update(self._extract_via_patterns(text))
        filtered = self._filter_skills(skills)
        logger.info(f"Extracted {len(filtered)} skills")
        return filtered

    def _extract_via_ner(self, text: str) -> Set[str]:
        doc = self.nlp(text)
        skills = set()
        for ent in doc.ents:
            if ent.label_ in {"PRODUCT", "ORG", "NORP", "LANGUAGE", "WORK_OF_ART"}:
                skills.add(ent.text.strip())
        for chunk in doc.noun_chunks:
            text_lower = chunk.text.lower()
            if any(indicator in text_lower for indicator in
                   ["software", "tool", "framework", "language", "platform", "system"]):
                skills.add(chunk.text.strip())
        return skills

    def _extract_via_tfidf(self, text: str, top_n: int) -> Set[str]:
        try:
            vectorizer = TfidfVectorizer(
                max_features=top_n, ngram_range=(1, 3), stop_words='english',
                token_pattern=r'\b[A-Za-z][A-Za-z0-9\.\+\#\-]*\b'
            )
            tfidf_matrix = vectorizer.fit_transform([text])
            feature_names = vectorizer.get_feature_names_out()
            scores = tfidf_matrix.toarray()[0]
            top_indices = scores.argsort()[-top_n:][::-1]
            return {feature_names[i] for i in top_indices if scores[i] > 0.01}
        except:
            return set()

    def _extract_via_patterns(self, text: str) -> Set[str]:
        skills = set()
        skills.update(re.findall(r'\b([A-Z][a-z]+(?:[A-Z][a-z]+)*(?:\.[a-z]+)?)\b', text))
        skills.update(re.findall(r'\b([A-Z]{2,6})\b', text))
        potential = re.findall(r'\b([A-Za-z]+(?:\s+\d+(?:\.\d+)?)?)\b', text)
        skills.update([p for p in potential if any(c.isdigit() for c in p)])
        skills.update(re.findall(r'\b([A-Za-z]+-[A-Za-z]+(?:-[A-Za-z]+)?)\b', text))
        return skills

    def _filter_skills(self, skills: Set[str]) -> Set[str]:
        filtered = set()
        for skill in skills:
            skill_clean = skill.strip()
            skill_lower = skill_clean.lower()
            if (2 <= len(skill_clean) <= 50 and
                skill_lower not in self.common_stop_words and
                re.search(r'[A-Za-z]', skill_clean) and
                skill_lower not in {"the", "and", "for", "with", "from", "this", "that", "will", "can"}):
                filtered.add(skill_clean)
        return filtered

# ---------- JD Extraction ----------
async def load_jd_from_url(url: str, session: aiohttp.ClientSession, timeout: int = 30) -> str:
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    }
    async with session.get(url, headers=headers, timeout=timeout, allow_redirects=True) as r:
        r.raise_for_status()
        html_text = await r.text()

    soup = BeautifulSoup(html_text, "html.parser")
    linkedin_selectors = [
        {"name": "div", "attrs": {"class": "description__text"}},
        {"name": "div", "attrs": {"class": "show-more-less-html__markup"}},
        {"name": "section", "attrs": {"class": "description"}},
    ]
    generic_selectors = [
        {"name": "div", "attrs": {"class": re.compile(r"(?i)(job[-_]?description|description)")}},
        {"name": "section", "attrs": {"class": re.compile(r"(?i)(job|description)")}},
    ]

    all_selectors = linkedin_selectors + generic_selectors
    text = ""
    for sel in all_selectors:
        node = soup.find(sel.get("name"), sel.get("attrs"))
        if node:
            text = node.get_text(separator="\n", strip=True)
            if len(text.split()) >= 50:
                break

    if not text or len(text.split()) < 50:
        paragraphs = soup.find_all("p")
        text = "\n".join([p.get_text(strip=True) for p in paragraphs])

    return re.sub(r"\n{3,}", "\n\n", html.unescape(text)).strip()

def read_docx(path: str) -> str:
    doc = Document(path)
    paragraphs = []
    for p in doc.paragraphs:
        text = p.text.strip()
        if text:
            paragraphs.append(text)
    for table in doc.tables:
        for row in table.rows:
            for cell in row.cells:
                text = cell.text.strip()
                if text:
                    paragraphs.append(text)
    return "\n".join(paragraphs)

def extract_experience_years(text: str) -> Optional[int]:
    patterns = [
        r"(\d+)\+?\s*(?:years?|yrs?)[\s\w]*(?:of\s+)?(?:experience|exp)",
        r"(?:minimum|at least)\s+(\d+)\s*(?:years?|yrs?)",
        r"(\d+)(?:\s*-\s*(\d+))?\s*(?:years?|yrs?)",
    ]
    max_years = None
    for pattern in patterns:
        matches = re.findall(pattern, text, re.I)
        for match in matches:
            nums = [int(n) for n in match if n and n.isdigit()]
            if nums:
                year = max(nums)
                if max_years is None or year > max_years:
                    max_years = year
    return max_years

# ---------- Main Matcher ----------
class ResumeJDMatcher:
    def __init__(self, config: Config, openai_api_key: str):
        self.config = config
        self.openai_api_key = openai_api_key  # ✅ BUGFIX: actually stores key
        self.resume_text: Optional[str] = None
        self.resume_parser: Optional[ResumeParser] = None
        self.jd_text: Optional[str] = None
        self.jd_criteria: Optional[Dict] = None
        self.jd_skills: Set[str] = set()
        self.resume_skills: Set[str] = set()
        self.resume_chunks: List[str] = []
        self.index: Optional[faiss.Index] = None
        self.resume_embs: Optional[np.ndarray] = None
        self._aiohttp_session: Optional[aiohttp.ClientSession] = None
        self._embed_semaphore = asyncio.Semaphore(self.config.concurrent_embeddings)
        self.skills_extractor = DynamicSkillsExtractor(nlp)

    async def ensure_session(self) -> aiohttp.ClientSession:
        if self._aiohttp_session is None:
            self._aiohttp_session = aiohttp.ClientSession()
        return self._aiohttp_session

    async def close(self):
        if self._aiohttp_session:
            await self._aiohttp_session.close()

    async def load_resume(self, path_or_text: str) -> None:
        if path_or_text.endswith(".docx") and os.path.exists(path_or_text):
            self.resume_text = read_docx(path_or_text)
            logger.info(f"✅ Loaded .docx: {path_or_text}")
        else:
            self.resume_text = path_or_text

        if not self.resume_text.strip():
            raise ValueError("Empty resume")

        self.resume_parser = ResumeParser(self.resume_text, nlp)
        self.resume_skills = self.skills_extractor.extract_from_text(self.resume_text)
        logger.info(f"Resume: {len(self.resume_text.split())} words, {len(self.resume_skills)} skills")

    async def load_jd(self, url_or_text: str) -> None:
        session = await self.ensure_session()
        if url_or_text.startswith("http"):
            self.jd_text = await load_jd_from_url(url_or_text, session)
            logger.info(f"✅ JD from URL: {len(self.jd_text.split())} words")
        else:
            self.jd_text = url_or_text

        if not self.jd_text.strip():
            raise ValueError("Empty JD")

        self.jd_skills = self.skills_extractor.extract_from_text(self.jd_text)
        logger.info(f"JD: {len(self.jd_text.split())} words, {len(self.jd_skills)} skills")

    def extract_jd_criteria(self) -> None:
        doc = nlp(self.jd_text)
        title = ""
        for line in self.jd_text.split("\n")[:15]:
            line_clean = line.strip()
            if 3 <= len(line_clean.split()) <= 12 and not line_clean.endswith(":"):
                title = line_clean
                break

        level_keywords = ["senior", "junior", "lead", "principal", "staff", "manager"]
        level = next((token.text for sent in list(doc.sents)[:20]
                     for token in sent if token.lower_ in level_keywords), "")

        exp_years = extract_experience_years(self.jd_text)
        qual_keywords = ["bachelor", "master", "phd", "degree", "certification"]
        quals = [sent.text.strip() for sent in doc.sents
                if any(kw in sent.text.lower() for kw in qual_keywords)][:10]

        action_verbs = {"develop", "design", "lead", "manage", "implement", "build"}
        resps = [sent.text.strip() for sent in doc.sents
                if any(token.lemma_ in action_verbs for token in sent)][:15]

        domain_keywords = ["healthcare", "medical", "dentistry", "dental", "finance", "fintech", "ecommerce", "saas", "cloud", "ai", "machine learning", "data science", "education", "retail", "human resources", "hr", "marketing", "sales", "engineering", "software development", "manufacturing", "law", "legal", "hospitality", "energy", "environment", "government", "media", "entertainment", "transportation", "logistics"]
        domain = [kw for kw in domain_keywords if kw in self.jd_text.lower()]

        self.jd_criteria = {
            "title": title, "level": level, "experience_years": exp_years,
            "skills": sorted(list(self.jd_skills))[:50],
            "qualifications": quals, "responsibilities": resps, "domain": domain
        }

    def chunk_resume(self) -> None:
        doc = nlp(self.resume_text)
        chunks = []
        current_chunk = []
        current_words = 0

        for sent in doc.sents:
            sent_text = sent.text.strip()
            sent_words = len(sent_text.split())
            if current_words + sent_words > self.config.chunk_size and current_chunk:
                chunk_text = " ".join(current_chunk)
                chunks.append(chunk_text)
                current_chunk = []
                current_words = 0
            current_chunk.append(sent_text)
            current_words += sent_words

        if current_chunk:
            chunks.append(" ".join(current_chunk))

        self.resume_chunks = chunks
        logger.info(f"Chunked into {len(chunks)} segments")

    async def _openai_embedding(self, text: str) -> np.ndarray:
        session = await self.ensure_session()
        endpoint = "https://api.openai.com/v1/embeddings"
        headers = {"Authorization": f"Bearer {self.openai_api_key}", "Content-Type": "application/json"}
        payload = {"input": text[:8000], "model": self.config.embed_model}

        async with session.post(endpoint, headers=headers, json=payload,
                               timeout=self.config.api_timeout) as r:
            data = await r.json()
        vec = np.array(data["data"][0]["embedding"], dtype=np.float32)
        norm = np.linalg.norm(vec)
        if norm > 0:
            vec /= norm
        return vec

    async def embed_resume(self) -> None:
        if not self.resume_chunks:
            self.chunk_resume()

        tasks = [self._openai_embedding(chunk) for chunk in self.resume_chunks]
        embeddings = []
        for task in tqdm(asyncio.as_completed(tasks), total=len(tasks), desc="Embedding"):
            embeddings.append(await task)

        self.resume_embs = np.vstack(embeddings).astype("float32")
        dim = self.resume_embs.shape[1]
        self.index = faiss.IndexFlatIP(dim)
        self.index.add(self.resume_embs)
        logger.info(f"FAISS: {self.resume_embs.shape[0]} vectors")

    async def retrieve_top_k(self, query_text: str, k: int) -> List[Dict]:
        query_vec = await self._openai_embedding(query_text)
        norm = np.linalg.norm(query_vec)
        if norm > 0:
            query_vec /= norm
        sims, indices = self.index.search(np.expand_dims(query_vec, 0), k)
        return [{"idx": int(idx), "similarity": float(sim), "text": self.resume_chunks[idx]}
                for sim, idx in zip(sims[0], indices[0]) if idx >= 0]

    def create_criterion_query(self, criterion_key: str) -> str:
        criteria = self.jd_criteria or {}
        queries = {
            "title_match": f"Job title: {criteria.get('title', 'Unknown')}. Relevant roles.",
            "level_seniority": f"Seniority: {criteria.get('level', 'Not specified')}. Leadership.",
            "experience_years": f"Required: {criteria.get('experience_years', 0)} years experience.",
            "technical_skills": f"Skills: {', '.join(list(self.jd_skills)[:20])}",
            "soft_skills": "Communication, leadership, teamwork, problem-solving.",
            "qualifications": f"Education: {'; '.join(criteria.get('qualifications', [])[:3])}",
            "responsibilities_alignment": f"Responsibilities: {' '.join(criteria.get('responsibilities', [])[:5])}",
            "domain_expertise": f"Domain: {', '.join(criteria.get('domain', ['general']))}"
        }
        return queries.get(criterion_key, self.jd_text[:500])

    def _get_enhanced_cot_prompt(self, criterion: str, query: str, passages: str) -> str:
        skill_overlap = len(self.jd_skills & self.resume_skills)
        skill_overlap_pct = (skill_overlap / len(self.jd_skills) * 100) if self.jd_skills else 0

        return f"""EXPERT RECRUITER EVALUATION - STRICT JD ALIGNMENT

You must evaluate how well the resume evidence aligns with the Job Description (JD) requirements for this criterion. Score ONLY based on direct matches to the JD's specific needs. Ignore any resume strengths that do not relate to the JD. Be extremely strict: if the resume does not demonstrate expertise in the JD's domain or field, score very low, even if the resume shows strong skills in unrelated areas.

CRITERION: {criterion}
QUERY: {query}

RESUME EVIDENCE:
{passages}


EVALUATE:
1. For each passage, extract evidence relevant ONLY to the JD (direct/transferable/gaps). Be strict: only evidence directly matching JD gets high passage scores; transferable to JD medium; unrelated or gaps low.
2. Assess relevance to JD, recency, and depth.
3. Synthesize holistically for overall JD match.

OUTPUT:
PASSAGE_ANALYSIS: [for each passage]
- Evidence: [specific facts relevant to JD]
- Reasoning: [why it matches or doesn't match the JD]
PASSAGE_RELEVANCE_SCORE: [0-100]


CONTEXT:
- JD Skills: {len(self.jd_skills)} | Resume Skills: {len(self.resume_skills)}
- Overlap: {skill_overlap} ({skill_overlap_pct:.1f}%)
- Exp Required: {self.jd_criteria.get('experience_years', 'N/A')} years

**SCORE THIS CRITERION (0-100) BASED ON JD ALIGNMENT:**
90-100: Exceeds JD requirements ✓ (perfect direct match, exceeds JD requirements)
75-89: Strong match to JD ✓ (strong direct match, minor gaps)
60-74: Adequate for JD ✓ (acceptable with some transferable skills to JD)
40-59: Partial match to JD ⚠ (partial match, significant gaps)
20-39: Weak match to JD ⚠ (weak relevance to JD, major mismatches)
0-19: No fit for JD ✗ (no relevance to JD or complete mismatch)

The FINAL_CRITERION_SCORE is a holistic assessment of JD match, not necessarily an average of passage scores. It can be lower if overall alignment is poor despite some weakly relevant passages.

OUTPUT EXACTLY:
FINAL_CRITERION_SCORE: [0-100]

RECOMMENDATIONS:
- [3-5 bullet points on improvements or strengths, tailored to this criterion and JD. If high score, highlight strengths aligning with JD; if low, suggest specific ways to improve resume or skills for better JD fit.]

JD REFERENCE:
{self.jd_text[:1500]}"""

# ---------- LLM Helpers ----------
async def _openai_chat_with_retry(session: aiohttp.ClientSession, api_key: str, messages: list,
                                  model: str, timeout: int = 120, retries: int = 3,
                                  temperature: float = 0.1, max_tokens: int = 1024) -> dict:
    endpoint = "https://api.openai.com/v1/chat/completions"
    headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}
    payload = {"model": model, "messages": messages, "temperature": temperature, "max_tokens": max_tokens}

    for attempt in range(retries):
        try:
            async with session.post(endpoint, headers=headers, json=payload, timeout=timeout) as r:
                if r.status != 200:
                    raise RuntimeError(f"API {r.status}")
                return await r.json()
        except Exception as e:
            if attempt == retries - 1:
                raise
            await asyncio.sleep(2 ** attempt)

def parse_final_score_from_text(text: str) -> Optional[int]:
    m = re.search(r"FINAL_CRITERION_SCORE\s*:\s*(\d{1,3})", text)
    if m:
        return max(0, min(100, int(m.group(1))))
    m2 = re.search(r"\b([0-9]{1,3})\b", text)
    if m2:
        s = int(m2.group(1))
        if 0 <= s <= 100:
            return s
    return None

# ---------- 🎯 MAIN FLOW ----------
async def main_flow(resume_path_or_text: Optional[str] = None,
                    jd_url_or_text: Optional[str] = None,
                    config: Optional[Config] = None,
                    openai_api_key: Optional[str] = None,
                    show_verbose: bool = True) -> Dict[str, Any]:

    if config is None:
        config = Config()
    openai_api_key = get_openai_api_key_interactive(openai_api_key)
    matcher = ResumeJDMatcher(config, openai_api_key)

    try:
        session = await matcher.ensure_session()

        # 🚀 INTERACTIVE INPUTS
        if resume_path_or_text is None:
            resume_path_or_text = prompt_for_resume_input()
        if jd_url_or_text is None:
            jd_url_or_text = prompt_for_jd_input()

        await matcher.load_resume(resume_path_or_text)
        await matcher.load_jd(jd_url_or_text)
        matcher.extract_jd_criteria()
        await matcher.embed_resume()

        criterion_results = {}
        for crit in matcher.config.criteria_keys:
            query = matcher.create_criterion_query(crit)
            passages = await matcher.retrieve_top_k(query, matcher.config.top_k_retrieve)

            passages_text = "\n\n".join([f"Passage {i+1}: {p['text']}"
                                       for i, p in enumerate(passages)])
            cot_prompt = matcher._get_enhanced_cot_prompt(crit, query, passages_text)

            messages = [{"role": "system", "content": "Expert recruiter."},
                       {"role": "user", "content": cot_prompt}]

            chat_resp = await _openai_chat_with_retry(session, openai_api_key, messages,
                                                    matcher.config.llm_model,
                                                    max_tokens=matcher.config.max_new_tokens)
            llm_text = chat_resp["choices"][0]["message"]["content"]
            llm_score = parse_final_score_from_text(llm_text)

            avg_sim = np.mean([p["similarity"] for p in passages]) * 100 if passages else 0
            emb_score = float(avg_sim)
            combined_score = (config.weight_llm * (llm_score if llm_score is not None else emb_score) + config.weight_emb * emb_score)
            combined_score = max(0, min(100, combined_score))

            criterion_results[crit] = {
                "llm_score": llm_score, "emb_score": round(emb_score, 2),
                "combined_score": round(combined_score, 2),
                "explanation": llm_text
            }

            if show_verbose:
                logger.info(f"{crit}: {combined_score:.1f} (LLM:{llm_score}, Emb:{emb_score:.1f})")

        overall_score = np.mean([v["combined_score"] for v in criterion_results.values()])

        result = {
            "overall_score": round(overall_score, 2),
            "criteria": criterion_results,
            "jd_criteria": matcher.jd_criteria,
            "resume_skills": sorted(list(matcher.resume_skills))[:50],
            "jd_skills": sorted(list(matcher.jd_skills))[:50]
        }

        print(f"\n🎉 FINAL SCORE: {result['overall_score']:.1f}/100")
        for k, v in result["criteria"].items():
            print(f"  {k}: LLM {v['llm_score']}, EMB {v['emb_score']:.1f}, Average {v['combined_score']:.1f}")
            print(f"    Explanation and Recommendations:\n{v['explanation']}\n")

        print("\nSummary Table:")
        print("| Criterion | LLM | EMB | Average |")
        print("|-----------|-----|-----|---------|")
        for k, v in result["criteria"].items():
            print(f"| {k} | {v['llm_score']} | {v['emb_score']:.1f} | {v['combined_score']:.1f} |")

        # Collect all recommendations
        all_recommendations = []
        for crit, res in criterion_results.items():
            expl = res['explanation']
            rec_start = expl.find("RECOMMENDATIONS:")
            if rec_start != -1:
                rec_text = expl[rec_start + len("RECOMMENDATIONS:"):].strip()
                bullets = re.findall(r'-\s*(.+)', rec_text)
                all_recommendations.extend([(crit, bullet) for bullet in bullets])

        aggregated_recs = "\n".join([f"{crit}: {rec}" for crit, rec in all_recommendations])

        # Update resume
        prompt_resume_update = f"""Based on the following recommendations to improve the resume for the JD, update the resume text accordingly.
Do not add any new information not present in the original resume. Only rephrase, reorganize, or highlight existing content to better align with the JD as per recommendations.
Original Resume:
{matcher.resume_text}
JD:
{matcher.jd_text}
Recommendations:
{aggregated_recs}
Output the updated resume text.
"""
        messages_resume = [{"role": "system", "content": "Expert resume writer."}, {"role": "user", "content": prompt_resume_update}]
        resp_resume = await _openai_chat_with_retry(session, openai_api_key, messages_resume, matcher.config.llm_model, max_tokens=2000)
        updated_resume_text = resp_resume["choices"][0]["message"]["content"]

        updated_resume_doc = Document()
        for line in updated_resume_text.split("\n"):
            updated_resume_doc.add_paragraph(line)
        updated_resume_path = "updated_resume.docx"
        updated_resume_doc.save(updated_resume_path)

        # Create cover letter
        prompt_cover = f"""Write a professional cover letter for the job based on the resume, JD, and recommendations to highlight strengths and address any gaps.
Resume: {matcher.resume_text}
JD: {matcher.jd_text}
Recommendations: {aggregated_recs}
Output the cover letter text.
"""
        messages_cover = [{"role": "system", "content": "Expert cover letter writer."}, {"role": "user", "content": prompt_cover}]
        resp_cover = await _openai_chat_with_retry(session, openai_api_key, messages_cover, matcher.config.llm_model, max_tokens=2000)
        cover_letter_text = resp_cover["choices"][0]["message"]["content"]

        cover_doc = Document()
        for line in cover_letter_text.split("\n"):
            cover_doc.add_paragraph(line)
        cover_path = "cover_letter.docx"
        cover_doc.save(cover_path)

        # Download handling
        def download_file(file_path):
            if "google.colab" in sys.modules:
                from google.colab import files
                files.download(file_path)
            elif "ipykernel" in sys.modules:
                print(f"File saved for download: {file_path}")
            else:
                print(f"File saved: {file_path}")

        print("\nDownloading updated resume...")
        download_file(updated_resume_path)
        print("\nDownloading cover letter...")
        download_file(cover_path)

        return result

    finally:
        await matcher.close()

# ---------- RUN (portable: script + notebook friendly) ----------
if __name__ == "__main__":
    try:
        # Try the simple, normal case first
        try:
            asyncio.run(main_flow())
        except RuntimeError as re:
            # Happens when running inside an existing event loop (e.g. Jupyter/Colab)
            logger.warning("Detected running event loop; attempting notebook-friendly execution.")
            try:
                # Prefer using nest_asyncio if available
                import nest_asyncio
                nest_asyncio.apply()
                loop = asyncio.get_event_loop()
                loop.run_until_complete(main_flow())
            except Exception:
                # If nest_asyncio is not available or something else failed,
                # provide clear instruction for notebook execution.
                logger.error("Could not auto-run inside the existing event loop.")
                logger.info("If you're in a Jupyter/Colab notebook, run: `await main_flow()`")
                raise
    except Exception as e:
        logger.exception(f"Failed: {e}")
```
